# Partie cours

## I - Présentation du problème

Le mot advection signifie **transport par le mouvement**. 

On cherche dans cette leçon une fonction du temps où l'inconnue $u(t)$ n'est plus un nombre (comme dans la première leçon) ou un vecteur de $\mathbb{R}^2$ (comme pour l'oscillateur harmonique). 

A un temps $ t $ fixé, l'inconnue $ u(t) $ est elle-même une fonction de l'espace $x$. En d'autres termes, l'inconnue
est une fonction $u(x, \, t )$ de deux variables. Physiquement, $u$ correspond à la quantité que l'on souhaite transporter (ex : un polluant, une densité, ...).  

L'équation d'advection n'est plus une équation différentielle, mais une équation dite **aux dérivées partielles**.

On se donne un nombre réel $ \, a \, $ qui, du point de vue de la physique est homogène au rapport d'une grandeur spatiale divisée par une grandeur temporelle : $ \, a \, $ est une **vitesse**.   
La fonction inconnue $ \, u(x, \, t) \, $ satisfait l'équation d'advection (1D), qui s'écrit :
 
$$ (1) \quad   \, \displaystyle  {{\partial u}\over{\partial t}} + a \, {{\partial u}\over{\partial x}}
= 0 \,, \quad x \in \mathbb{R} , \,\, t > 0 \, . $$ 

Pour fermer le système, on se donne une **condition initiale** en tout point d'espace:  

$$ (2) \quad   \, \displaystyle  u(x, \, 0) = u_0 (x)  \,, \quad x \in  \mathbb{R} \, . $$ 

**Remarque** : Dans le cas de la mécanique des fluides, lorsque l'on souhaite transporter un fluide de vitesse $\bf{u}$ (en dimension $d=1,2,3$), l'équation de transport s'écrit :
$$ \dfrac{\partial {\bf{u}}}{\partial t} + ({\bf{u}} \cdot \nabla) {\bf{u}} = 0 \,, \quad {\bf{u}} \in \mathbb{R}^d , \,\, t > 0$$

Dans ce cas, la quantité transportée et la vitesse à laquelle on la transporte sont identiques : il s'agit du champ de vitesse ${\bf{u}}$ lui-même. Le membre de gauche de cette équation ci-dessus correspond exactement au terme de transport dans l'équation de conservation de la quantité de mouvement des équations de **Navier-Stokes**.

Revenons au cas de l'équation de transport 1D (équation $(1))$. Afin de résoudre numériquement cette équation, on considère un domaine spatial fini $[0,L]$ ce qui nécessite pour tout pas de temps le choix de **conditions aux limites** au bord du domaine spatial.

Dans le cadre de ce cours, on choisit des **conditions aux limites périodiques** : $u(L,t) = u(0,t), \quad t \geq 0.$

## II - Solution exacte : méthode des caractéristiques

On peut construire une solution analytique du problème continu $(1) (2)$. 

On se donne $ \, y \in \mathbb{R} \, $ et on pose $ \, \, \displaystyle v(t) = u( \underbrace{a\, t + y}_x , \, t ) $. 

Alors par dérivation des fonctions composées pour des fonctions de deux variables, on a :

$$ \,  \displaystyle {{{\rm d} v}\over{{\rm d}t}} =  
 \,  \displaystyle {{{\rm d} x}\over{{\rm d}t}} \,  {{\partial u}\over{\partial x}}  + 
{{\partial u}\over{\partial t}} \, =  
 \,  \displaystyle \, a \, {{\partial u}\over{\partial x}}  + 
{{\partial u}\over{\partial t}} \,\, $$  
et cette expression est nulle si la fonction $ \, u(x, \, t) \, $ est solution de l'équation $(1)$. La fonction $ \, v(t) \, $ est donc constante et est égale à sa valeur initiale lorsque  $ \, t = 0 $.


On a donc la suite d'égalités
 $\,\, \,  u( a\, t + y , \, t ) \, $ $ = v(t) = v(0) \, $ $ = u( y , \, 0 ) = u_0(y) $. 

Comme $ \, y \, $ est arbitraire, on peut aussi l'écrire $ \, y = x - a \, t $, 
avec $ \, x \, $ quelconque. 

La relation précédente entraîne  donc 
$$(3) \quad   \, \displaystyle  u(x, \, t) = u_0 ( x - a \, t ) \,, \quad x \in \mathbb{R} \,, \,\, t \geq 0 \, .   $$ 

On peut vérifier par un calcul élémentaire que la relation $(3)$ est bien solution d'une part de l'équation d'évolution $(1)$,  d'autre part de la condition initiale $(2)$. 


Le choix d'une solution initiale périodique de période $L$ 
$$ u_0(x)= u_0(x+L) $$

assure la périodicité de la solution :
$$\begin{array}{rcl}
u(x+L,t)=u_0(x+L-at)&=&u_0(x-at \,+L)\\
&=&u_0(x-at)=u(x,t), \quad x\in \mathbb{R},\quad t\geq 0
\end{array}
$$

On appelle **lignes caractéristiques** toutes les droites $x-at$ de pente $a$. 

**Bilan** : La solution de l'équation en un point $x$ donné de l'espace reste donc constante en temps le long de sa ligne caractéristique. En d'autres mots, la valeur de la solution en un point $(x,t)\in \mathbb {R} \times \mathbb {R} ^{+}$ peut être retrouvée en cherchant la valeur de la condition initiale $u_0$ **au pied de la ligne caractéristique** $x_0 = x - a t$ :

$$ \boxed{u(x,t)=u_{0}(x-a\,t)} \quad \forall \ (x,t)\in \mathbb {R} \times \mathbb {R} ^{+}$$

## III - Approximation numérique

La méthode des caractéristiques est exacte et très simple à utiliser, mais elle présente l'inconvénient de devoir connaître l'expression analytique de la solution initiale $u_0(x)$. Or en pratique, il arrive bien souvent que cela ne soit pas le cas. En effet, lorsqu'on souhaite résoudre numériquement une EDP, l'état initial de notre système physique (c'est à dire la condition initiale) ne nous est souvent pas donné sous forme continue (c'est à dire sous la forme d'une formule mathématique) mais bien souvent sous forme discrète, c'est à dire sous la forme d'un tableau de valeurs discrètes $u_0[0, \Delta x, 2\Delta x, \cdots, L]$ (récupéré dans un fichier de données, dans un programme, construit soi même ...).  

Dans ce cas là, on ne peut plus utiliser la méthode exacte des caractéristiques et il devient nécessaire de mettre en place un schéma numérique permettant de discrétiser l'équation de transport $(1)$ :

$$ (1) \quad   \, \displaystyle  {{\partial u}\over{\partial t}} + a \, {{\partial u}\over{\partial x}}
= 0 \,, \quad x \in \mathbb{R} , \,\, t > 0 \, . $$

### 1 - Une première tentative numérique : le schéma centré

On se donne un pas d'espace $ \, \Delta x > 0 \, $ et un pas de temps $ \,  \Delta t > 0  $. 
On approche la solution exacte $ \, u(j \, \Delta x , \, n \, \Delta t ) \, $ par le nombre $ \, u_j^n $, avec un indice en position inférieure pour décrire la variation en espace et un indice en position supérieure pour l'évolution temporelle:

$$ (4) \quad   \, \displaystyle  u_j^n \approx  u(j \, \Delta x , \, n \, \Delta t )  \, . $$ 

Pour discrétiser l'équation de transport $(1)$, il faut approcher la dérivée en temps $\dfrac{\partial u}{\partial t}$ ainsi que la dérivée en espace $\dfrac{\partial u}{\partial x}$.

* On approche la dérivée en espace par un schéma centré :  

**Construction d'un schéma centré** pour l'approximation d'une dérivée d'ordre 1 en espace :  

En un point $x$ et pour une valeur $h$ du pas de discrétisation tels que $u$ soit trois fois dérivable sur l’intervalle $[x - h, x + h]$, la formule de Taylor-Young conduit aux deux relations :
$${\displaystyle u(x+h)=u(x)+ h u'(x) + {\frac {h^2}{2!}}u''(x) + {\frac {h^{3}}{3!}}u^{(3)}(x)+h^{3}\varepsilon_{1}(x,h)}$$

$${\displaystyle u(x-h)=u(x)- h u'(x) + {\frac {h^2}{2!}}u''(x) - {\frac {h^{3}}{3!}}u^{(3)}(x) +h^{3}\varepsilon_{2}(x,h)} $$

où les deux fonctions $\varepsilon_i(x , h)$ convergent vers 0 avec $h$.

En soustrayant les développements précédents, ce qui revient à faire la moyenne des deux différences finies antérieure et postérieure à $u(x)$, on obtient :

$${\displaystyle {\frac {u(x+h)-u(x-h)}{2h}}=u'(x)+{\frac {h^{2}}{6}}u^{(3)}(x)+h^{2}\varepsilon_{3}(x,h)} $$

qui est une approximation de $u'(x)$ du 2nd ordre en $h$, que l'on pourrait aussi ré-écrire ainsi en posant $h=\Delta x$ :

$$u'(x) = \dfrac{\partial u}{\partial x}(x) \approx \displaystyle {\frac {u(x+\Delta x)-u(x-\Delta x)}{2 \Delta x}}$$

Ainsi, dans notre cas, l'approximation de la dérivée en espace de $u(x,t)$ par un schéma centré donne :

 $$  \displaystyle  {{\partial u}\over{\partial x}} (j \, \Delta x , \, n \, \Delta t ) 
\approx \displaystyle {\frac {u^n(j\Delta x + \Delta x)-u^n(j\Delta x-\Delta x)}{2 \Delta x}} = {{u_{j+1}^n - u_{j-1}^n }\over{2 \, \Delta x}}  \quad {\text{(approx. d'ordre 2)}}$$  

* Concernant la dérivée en temps, on l'approche par un schéma explicite : 

 $$  \,\,\, \displaystyle  {{\partial u}\over{\partial t}} (j \, \Delta x , \, n \, \Delta t ) 
\approx {{u_{j}^{n+1} - u_{j}^n }\over{\Delta t}}  \quad {\text{(approx. d'ordre 1)}}$$

On obtient de cette façon le schéma **centré en espace et explicite en temps** pour l'équation d'advection :

$$ (5) \quad  \, \displaystyle  {{u_{j}^{n+1} - u_{j}^n }\over{\Delta t}} + a \, 
{{u_{j+1}^n - u_{j-1}^n }\over{2 \, \Delta x}} = 0  \, . $$ 


Si le champ discret $ \, u_j^n \, $ est donné à l'instant discret $ \, n \,$ pour toutes
les valeurs de l'espace discret $ \, j $, le calcul de la solution approchée au nouvel instant 
ne pose aucune difficulté :
 $$ (6) \quad \boxed{ \displaystyle   u_{j}^{n+1} =  u_{j}^n  
- {{a \, \Delta t}\over{2 \, \Delta x}} \, \big( u_{j+1}^n - u_{j-1}^n \big)} $$ 



Le calcul réel s'effectuant sur une grille de taille finie, on a  $ \, 0 \leq j \leq J $ avec $J\, \Delta x = L.$

La condition aux limites périodique s'écrit :
$ \, u_J^n = u_0^n  $.  

#### Stabilité du schéma centré :

Le schéma centré en espace et explicite en temps $(6)$, est très sain du point de vue de sa conception en termes d'approximation des diverses dérivées partielles du problème. On dit qu'il est **consistant** avec l'équation aux dérivées partielles $(1)$. 

Mais l'expérience numérique  montre qu'après quelques itérations, 
l'approximation numérique $ \, u_j^n \, $ s'éloigne de la solution exacte voire explose.

On constate que le schéma centré est **instable**. Et ceci se produit quelle que soit la valeur (strictement positive) du nombre CFL défini, dans le cas de l'équation d'advection, par :

$$ \quad   \, \displaystyle  {\rm{CFL}} = {{a \, \Delta t}\over{\Delta x}}  \, . $$ 

### 2 - Le schéma décentré amont / aval

#### Cas où $ \, a > 0 $ : schéma décentré amont
On se place tout d'abord dans le cas où $ \, a > 0 $.

On approche désormais la dérivée en espace par un schéma décentré :    

**Construction du schéma décentré amont** :

En un point $x$ et pour une valeur $h$ du pas de discrétisation tels que $u$ soit trois fois dérivable sur l’intervalle $[x, x + 2h]$, la formule de Taylor-Young conduit à la relation :
$${\displaystyle u(x-h)=u(x)- h u'(x) + {\frac {h^2}{2!}}u''(x) - {\frac {h^{3}}{3!}}u^{(3)}(x) +h^{3}\varepsilon_(x,h)} $$

où la fonction $\varepsilon(x,h)$ converge vers 0 avec $h$. Par conséquent :

 $${\displaystyle {\frac {u(x)-u(x-h)}{h}}=u'(x)-{\frac {h}{2}}u''(x)+{\frac {h^{2}}{3!}}u^{(3)}(x)-h^{2}\varepsilon _{1}(x,h)} $$

correspond à une approximations de $u'(x)$ du 1er ordre en $h$. 

Ainsi, dans notre cas, l'approximation de la dérivée en espace par un schéma décentré amont donne :

 $$  \displaystyle  {{\partial u}\over{\partial x}} (j \, \Delta x , \, n \, \Delta t ) 
\approx {{u_{j}^n - u_{j-1}^n }\over{\Delta x}}  \quad {\text{(approx. d'ordre 1)}}$$  




On en déduit ainsi l'expression du **schéma décentré amont pour l'équation d'advection** : 

$$ (7) \quad   \, \displaystyle  {{u_j^{n+1} - u_j^{n}}\over{\Delta t }} + a \, 
{{u_j^{n} - u_{j-1}^{n} }  \over{\Delta x }} = 0 \, $$ 

que l'on peut ré-écrire sous la forme :

$$ (8) \quad   \,  \boxed{\displaystyle  u_j^{n+1} =   u_j^{n} - {{a \, \Delta t }\over{\Delta x }} \, 
\big( u_j^{n} - u_{j-1}^{n} \big) \ }$$ 


On a remplacé la dérivée partielle $ \, \displaystyle {{\partial u}\over{\partial t}} \, $
par le schéma explicite en temps  $ \,  \displaystyle  {{u_j^{n+1} - u_j^{n}}\over{\Delta t }} \, $ 
et la dérivée partielle $ \,  \displaystyle {{\partial u}\over{\partial x}} \, $ par le schéma 
décentré à gauche   $ \,  \displaystyle {{u_j^{n} - u_{j-1}^{n} }  \over{\Delta x }} \, $: 
on va chercher l'information dans la direction amont relativement à la vitesse $ \, a > 0$, ainsi 
que le suggère la méthode des caractéristiques $(3)$. 

#### Cas où $ \, a < 0 $ : schéma décentré aval

Dans le cas $ \, a < 0 $, on doit simplement remplacer la dérivée partielle en espace 
 $ \,  \displaystyle {{\partial u}\over{\partial x}} \, $  par le **schéma décentré aval** $ \,  \displaystyle {{u_{j+1}^{n} - u_{j}^{n} }  \over{\Delta x }} $. 

## IV - Analyse de stabilité de Von Neumann


La relation $(8)$ est explicite. Supposons  qu'à l'instant $ \, n \, \Delta t $, la fonction discrète $ \, u_j^n \, $ est une onde de vecteur d'onde $ \, K \, $:

$$  \, \, \displaystyle  u_j^{n} = \widehat{u}_K^n \, \exp ( i \, K \, j \, \Delta x ) $$

Alors après un calcul élémentaire, compte tenu de $(8)$ et de la définition du nombre CFL :

$$(9) \quad   \, \displaystyle  \sigma = {{a \, \Delta t}\over{\Delta x}} $$ 

on a : 
$$(10) \quad   \, \displaystyle  u_j^{n+1} = g (\sigma ,\, \xi) \,  u_j^{n} \,, \quad 
g(\sigma ,\, \xi) = 1 - \sigma \, \big( 1 - \exp (-i \, \xi ) \big) \,, \quad \xi = K \, \Delta x \, .$$

La fonction discrète $ \, u_j^{n+1} \, $ est donc aussi une onde de vecteur d'onde $ \, K \, $:
$  \, \, \displaystyle  u_j^{n+1} = \widehat{u}_K^{n+1} \, \exp ( i \, K \, j \, \Delta x ) $ et 

$$
\widehat{u}_K^{n+1}=g (\sigma ,\, \xi) \, \widehat{u}_K^{n}
$$

Le schéma décentré se réduit  dans ce cas à une simple suite géométrique, de raison 
égale au coefficient  d'amplification $ \, g (\sigma ,\, \xi) $. 

Le schéma est stable si cette suite géométrique ne tend pas vers l'infini en module
quand l'entier $ \, n \, $ tend vers l'infini, 
c'est à dire si pour tout nombre d'onde $ \, \xi $, le module de 
$ \, g (\sigma ,\, \xi) \, $ est inférieur ou égal à 1:
$$(11) \quad   \, \displaystyle  | g (\sigma ,\, \xi) | \leq 1  \,, \quad \forall  \, \xi \in {\rm R} \, .$$

La condition $(11)$ est équivalente à   

$$ 
(12) \quad   \, \displaystyle 0 \leq \sigma = {\text{CFL}} = {{a \, \Delta t}\over{\Delta x}} \leq 1 \, . $$ 


* Si le pas de temps $ \, \Delta t \, $ ne satisfait pas la condition $(12)$, le schéma est instable et donc ne converge pas.

* Si le pas de temps $ \, \Delta t \, $ n'est pas trop grand, c'est à dire s'il satisfait $(12)$ :
 $$\, \displaystyle \Delta t \leq {{\Delta x}\over{a}}  \Longleftrightarrow \Delta t = \dfrac{{\text{CFL}} \cdot \Delta x}{a} \ {\text{avec CFL}} \leq 1$$ 
 alors le schéma décentré est stable.

## Partie A : Solution exacte (solution périodique sinusoïdale)

**1)** 
* La fonction python `tabx` ci-dessous a pour rôle de remplir le tableau des positions $x_j$ discrètes (`param.x`).  
Ces positions $x_j$ sont au nombre de $J + 1$ et vont de 0 jusqu'à $L$ par pas de $\Delta x$.
* Pour cette étude on propose de se donner comme solution initiale périodique la fonction $u_0(x)= \sin(2 \pi \displaystyle {x \over L})$. D'après la méthode des caractéristiques, on sait que la solution exacte $u$ en chaque point $x$ et chaque instant $t$ est égale à :$$u(x, \, t) = u_0 ( x - a \, t )$$

Celle-ci est implémentée dans la fonction python `sinus` ci-dessous : pour une position $x$ et un temps $t$ elle renvoie : 
$$(x,t) \mapsto u_0(x-at) = \sin \big(2 \pi \displaystyle {(x-at) \over L} \big )$$
D'où la fonction python `sol_exacte` donnée ci-après, faisant directement appel au programme `sinus`.

```python
# Import des packages python
%matplotlib notebook
%matplotlib inline

import time
import numpy as np
import scipy.linalg as sp_linalg
import scipy.optimize as sp_optimize
import matplotlib.animation as animation
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from IPython import display
```

```python
class Param(object) : 
    def __init__ (self):
        # Parametre(s) du problème d'advection
        self.a = 1.                    # vitesse d'advection (constante)
        
        # Paramètres de discrétisation spatiale
        self.L = 1.                    # fin du domaine spacial [0,L]
        self.J = 20                    # nombre de pas d'espace
        self.dx = 0.                   # pas d'espace
        self.x = np.zeros(self.J+1)    # tableau des positions x le long du segment [0,L]
        
        # Paramètres de discrétisation temporelle
        self.T = 1.                    # temps final de la simulation
        self.n = 0                     # nombre d'itérations en temps
        self.dt = 0.                   # pas de temps
        self.tn = 0.                   # valeur du temps tn courant (attention c'est un reel, pas un tableau)
        self.cfl = 0. 

        # tableaux numpy pour les solutions (solution exacte + solution approchee au temps courant et au temps précédent)
        self.solution_exacte = np.zeros(self.J+1)       # solution exacte pour tout x au temps tn 
        self.solution_courante = np.zeros(self.J+1)     # u[:,tn+1] : solution approchee u pour tout x au temps tn+1
        self.solution_prec = np.zeros(self.J+1)         # u[:,tn] : solution approchee u pour tout x au temps tn
        
        # Fonctions python
        self.function = sinus
        self.schema = schema_decentre_amont
```

```python
def tabx(param) :
    param.x = np.linspace(0, param.L, param.J + 1)
```

```python
def sinus(param) :
    s = np.sin(2. * np.pi * (param.x - param.a * param.tn) / param.L)  # s est un tableau numpy de taille J+1 
    return s
```

```python
def sol_exacte(param) :
    param.solution_exacte = np.zeros(param.J + 1)
    param.solution_exacte = sinus(param)
    return param.solution_exacte
```

## Partie B : Solution approchée avec le schéma décentré amont / upwind

**2)**  Complétez la fonction python `schema_decentre_amont` ci-dessous qui implémente le schéma décentré amont pour la résolution de l'équation de transport :

$$  (8) \qquad \displaystyle   u_{j}^{n+1} =  u_{j}^n  - {{a \, \Delta t}\over{\, \Delta x}} \, \big( u_{j}^n - u_{j-1}^n \big)$$

**Remarque** : attention aux valeurs de la solution $u$ aux points $x$ situés sur les bords (i.e. $x=0$ et $x=L$). Rappel, les conditions de bord choisies sont des conditions de bord *périodiques*.

```python
def schema_decentre_amont(param):
    sigma = param.a * param.dt / param.dx 
    #sigma = param.cfl
    param.solution_courante[1:param.J+1] = param.solution_prec[1:param.J+1] - sigma * (param.solution_prec[1:param.J+1] - param.solution_prec[0:param.J])
    param.solution_courante[0] = param.solution_prec[0] - sigma * (param.solution_prec[0] - param.solution_prec[param.J-1])
    return param.solution_courante
```

**3)** Les implémentations concernant le calcul de la solution exacte et de la solution approchée à un temps $t$ quelconque sont faites. Il ne reste plus qu'à implémenter les 2 étapes primordiales de l'algorithme de résolution du problème de transport à savoir :
* l'initalisation du problème pour $t=0$ : pour cela complétez la fonction python `initialisation` ci-dessous dont le rôle est d'initialiser `param.x` (tableau des coordonnées en $x$), ainsi que `param.solution_exacte` d'un côté et `param.solution_prec` et `param.solution_courante` de l'autre, à leurs valeurs initiales respectives.   
Question : avec quelles valeurs proposez-vous d'initialiser solution_prec et solution_courante  ?  

* la boucle en temps pour $0<t\leq T$ : pour cela complétez la fonction python `run` ci-dessous, qui pour une iteration $n$ donnée, met à jour la valeur du temps associé `param.tn` et met à jour `param.solution_exacte` pour ce temps là ainsi que `param.solution_prec` et `param.solution_courante`.

```python
def initialisation(param):
    tabx(param)

    # temps initial
    param.tn = 0.0

    param.solution_exacte = sol_exacte(param)
    param.solution_courante = np.zeros(param.J + 1)
    param.solution_prec = np.zeros(param.J + 1)
    
    param.solution_prec = param.solution_exacte.copy()
    param.solution_courante = param.solution_exacte.copy()
```

```python
def run(iter_n, param) :
    param.tn = iter_n * param.dt
    param.solution_exacte = sol_exacte(param)
    param.solution_courante = param.schema(param)
    param.solution_prec = param.solution_courante.copy()
```

**4)** Mettez en place les paramètres de votre simulation et appelez une des deux fonctions python données ci-dessous$^{\boldsymbol{*}}$. Ces fonctions python résolvent l'algorithme (initialisation + boucle en temps) tout en traçant simultanémant de façon dynamique les solutions exactes et approchées.

Vous baserez la définition de votre pas de temps $\Delta t$ sur la définition suivante, faisant intervenir le nombre CFL :

$$ \Delta t  = \dfrac{{\rm{CFL}} \cdot \Delta x }{a}$$ 

On fera une première simulation pour une petite valeur du coefficient CFL (par exemple CFL = 0.1-0.2) puis pour une valeur un peu plus grande (par exemple CFL = 0.6-07).   
$\Longrightarrow$ Qu'observez-vous ?

$^{\boldsymbol{*}}$ Les deux fonctions python au choix :
* `run_and_animate_fast_exec` : calcul des solutions + affichage animé avec **temps d'execution très rapide** mais **pas d'interaction** avec le graphe produit

## Partie C : Solution approchée avec schéma centré

**5)**  Complétez la fonction python `schema_centre` ci-dessous qui implémente le schéma centré pour la résolution de l'équation de transport :

$$  (6) \qquad \displaystyle   u_{j}^{n+1} =  u_{j}^n  - {{a \, \Delta t}\over{2 \, \Delta x}} \, \big( u_{j+1}^n - u_{j-1}^n \big)$$
où ici :
* $u_{j}^{n+1}$ $\longleftrightarrow$ attribut `solution_courante` de l'objet `param`
* $u_{j}^{n}$ $\longleftrightarrow$ attribut `solution_prec` de l'objet `param` 

Puis testez-le avec des simulations où vous testerez plusieurs valeurs de CFL.  
$\Longrightarrow$ Qu'observez-vous ?

```python
def schema_centre(param):
    sigmaS2 = param.a * param.dt / param.dx /2
    #sigmaS2 = param.cfl/2
    param.solution_courante[1:param.J+1] = param.solution_prec[1:param.J] - sigmaS2 * (param.solution_prec[1:param.J+1] - param.solution_prec[0:param.J-1])
    param.solution_courante[0] = param.solution_prec[0] - sigmaS2 * (param.solution_prec[0] - param.solution_prec[param.J-1])
    return param.solution_courante
```

## Partie D : Etude de convergence 

**6)** Faire l'analyse de convergence (calcul d'erreurs puis tracé/calcul de l'ordre de convergence) pour 8 simulations réalisées avec le **schéma décentré amont**. 

Pour les 8 simulations, on partira de $J = 16$ (simu 1) et on ira jusqu'à $J = 2048$ (simu 8) en multipliant à chaque fois $J$ par 2. On gardera la **même CFL** pour les 8 simulations.  

$\Longrightarrow$ Compléter les différents blocs de la cellule ci-dessous :

```python
param = Param()
param.J = 8
param.cfl = 0.75
param.T = 1.0
param.schema = schema_decentre_amont
Nmesh = 8
err = np.zeros(Nmesh)
pas = np.zeros(Nmesh)
for imesh in range(0,Nmesh) :
    param.J = param.J * 2
    param.dx = param.L / param.J
    param.dt = param.cfl / param.a * param.dx
    param.n = int(param.T / param.dt)
    print('J=', param.J, 'n=', param.n) 
    initialisation(param)
    for nt in range(1,param.n+1):
        run(nt,param)
    diff = np.zeros(param.J+1)
    diff[0:param.J+1] = param.solution_exacte[0:param.J+1] - param.solution_courante[0:param.J+1]
    absdiff = abs(diff)
    err[imesh] = np.log(max(absdiff)) / np.log(2)
    pas[imesh] = np.log(param.J) / np.log(2)
```

Appeler les fonctions python `PlotConv` (=tracé la courbe de convergence) et `OrdreConv` (=calcul de l'ordre de convergence) données ci-dessous et parfaitement semblables à celles déjà implémentées dans le TP2.

$\Longrightarrow$ Quel est l'ordre du schéma ? Cela vous semble-t-il cohérent avec le cours ?

```python
def PlotConv(pas,err) :
    '''
    Graphique de la convergence
    '''
    fig, ax = plt.subplots(figsize=(5,5))
    ax.set_title( 'convergence en maillage à CFL fixée' ) 
    ax.set_xlabel('logarithme base 2 du nombre d éléments')
    ax.set_ylabel('logarithme base 2 de l erreur')
    ax.plot (pas,err,'r*')   
    plt.draw()
    return
```

```python
def OrdreConv(pas, err, m) :
    '''
    Calcul de l'ordre de convergence
    par approximation des moindres carrés
    '''
    xm = np.mean(pas)        
    ym = np.mean(err)        
      
    x2m = np.var(pas)
    y2m = np.var(err)
    xym = np.dot(pas,err) / m
    alpha = np.cov(pas,err,bias=True)[0,1] / x2m
    
    return np.abs(alpha)
```

```python
# Tracé de la courbe de convergence
PlotConv(pas,err)
```

```python
# Calcul de l'ordre de convergence
alpha = OrdreConv(pas, err, Nmesh)
print ('Ordre de convergence du schema decentré amont =', alpha)  
```
