## II - Résolution numérique des EDO (suite)

### 3) Définition de l'ordre d'une méthode numérique

Les schémas d'Euler explicite et implicite semblent avoir la même précision. 

Nous allons calculer numériquement l'**ORDRE** de ces shémas.

On se donne le temps $T$ final et on définit l'erreur entre la solution numérique et la solution exacte au temps $T$, pour différentes valeurs de $\Delta t$ :

$$
err(\Delta t) = \big \vert \,\underbrace{u^N}_{\text{sol. numérique}} - \underbrace{u(N\, \Delta t)}_{\text{sol. exacte}}\, \big \vert \ , \qquad  \quad {\rm avec} \ \ T=N\,\Delta t.
$$

Le schéma **converge** si :

$$err(\Delta t) \longrightarrow 0 \quad {\rm quand} \quad \Delta t \longrightarrow 0$$

Et si de plus, on a :

$$err(\Delta t) = O(\Delta t^p),\quad p>0 $$ 

on dit que le schéma **converge à d'ordre $\boldsymbol{p}$**.

#### a) Représentation graphique de l'ordre d'une méthode numérique

Si $err(\Delta t) = O(\Delta t^p)$ et si le nombre total d'itérations est égal à $N=\displaystyle{{T}\over{ \Delta t}}= 2^q$ alors :

$$\ln(err) = \ln(\Delta t^p) = p \ln(\Delta t) = p \ln \left (\dfrac{T}{N} \right) = p \ (\ln(T) - \ln(N)) = -p \ln(N) + \underbrace{p \ln(T)}_{C(T)}$$

$$
 \begin{array}{rcl}\Longleftrightarrow \ln(err) &=& -p \ln(2^q) + C(T) \\
&=&  -p \ q \ln(2) + C(T) 
\end{array}
$$

Ainsi on a :
$$
 \begin{array}{rcl}\dfrac{\ln(err)}{\ln(2)} &=&  -p \, q + \tilde{C}(T) \\
&=&   -p \, \dfrac{\ln(N)}{\ln(2)} + \tilde{C}(T) \ \ \rm{(car \ N=2^q)}
\end{array}
$$

qui est l'équation d'une droite.

Ce qui signifie que si l'on trace le logarithme en base 2 de l'erreur (i.e. $\dfrac{\ln(err)}{\ln(2)}$) en fonction du logarithme en base 2 du nombre de pas de temps dans la discrétisation temporelle (i.e. $\dfrac{\ln(N)}{\ln(2)}$), alors la courbe obtenue sera une droite décroissante dont le coefficient directeur sera, en valeur absolue, $p$.   
$\rightarrow$ C'est ainsi que l'on pourra **lire graphiquement l'ordre de convergence $p$** d'une méthode numérique.

#### b) Détermination expérimentale de l'ordre d'une méthode numérique par régression linéaire

Cependant, une simple lecture graphique se révèlera souvent insuffisante pour déterminer précisément l'ordre numérique obtenu, car les points de la courbe de convergence obtenus par simulations numériques ne seront pas nécessairement alignés parfaitement le long d'une droite.   
Pour connaître quantitativement cet ordre, on fait une **regression linéaire**.

**Rappel sur la méthode de régression linéaire** : 

Principe : Le but de cette méthode est de déterminer une droite $D$ dont le graphe passe au plus près d'un nuage de $N$ points $(X_i,Y_i)_{i=1...N}$.   
Soit $D(X_i)$ la valeur de cette droite au point $X_i$, autrement dit $D(X_i) = \alpha X_i + \beta$. L'objectif est de trouver, par une approximation des moindres carrés, les valeurs $\alpha ,\, \beta$ qui minimisent la somme des écarts au carré entre les mesures $Y_i$ et les valeurs de la droite en $X_i$ (on note cette quantité $S$) :

$$ \min_{\alpha, \ \beta \ \in \ \mathbb{R}} f(\alpha, \beta) = \sum_{i=1}^N (Y_i - \alpha X_i - \beta)^2 $$

</center>

$\rightarrow$ Ces valeurs $\alpha$ et $\beta$ ne sont autres que le coefficient directeur et l'ordonnée à l'origine de la droite $D$ que l'on cherche.

On introduit les quantités centrées : $\tilde Y_i=Y_i - <Y>$ et $\tilde X_i=X_i - <X>$ avec :

$$<Y>=\dfrac{1}{N}\sum_{i=1}^N Y_i \quad <X>=\dfrac{1}{N}\sum_{i=1}^N X_i \quad \text{(moyennes)}$$

* La condition d'optimalité par rapport à $ \ \beta \ $ s'obtient en résolvant l'équation $\dfrac{\partial f}{\partial \beta}(\alpha, \beta) = 0$, ce qui donne après calcul :

$$ <Y> -\alpha <X> -\beta =0$$
$$ \Longleftrightarrow \beta \ = \ <Y> -\alpha <X> $$

et en remplaçant $\beta$ dans l'expression de la fonctionnelle $f$, cette dernière s'exprime maintenant seulement en fonction de $\alpha$ et le problème de minimisation devient alors :
$$
\min_{\alpha \in \ \mathbb{R}} f(\alpha) =\sum_{i=1}^N (\tilde Y_i - \alpha \tilde X_i )^2
$$

* La condition d'optimalité par rapport à $ \ \alpha \ $ s'obtient en résolvant l'équation $f'(\alpha) = 0$, ce qui donne :

$$\sum_{i=1}^N \left (\tilde Y_i \tilde X_i- \alpha (\tilde X_i )^2 \right ) =0$$

On en déduit :
$$\boxed{ \alpha= \displaystyle{{\displaystyle\sum_{i=1}^N \tilde Y_i \tilde X_i}\over{\displaystyle\sum_{i=1}^N \tilde X_i ^2  }}}$$
ce qui s'écrit aussi (interprétation statistique de la régression linéaire) :

$$ \boxed{
\alpha= \dfrac{{\rm{Cov}}(X,Y)}{{\rm{Var}}(X)}}
$$


**Conclusion** : L'ordre de convergence $p$ d'une méthode numérique sera donc donné par la valeur absolue de $\alpha$, qui est le coefficient directeur de la droite de regression.

```python
# Import des packages python
import time
import numpy as np
import scipy.optimize
import matplotlib.pyplot as plt
```

```python
class Param (object) : 
    ''' 
    Creation d'une classe Param dont la fonction principale (__init__) 
    spécifie tous les attributs des objets qui seront issus de cette classe Param.
    Ces attributs correspondent à l'ensemble des paramètres 
    à manipuler dans les simulations numériques.
    Les valeurs/instances données ici à ces paramètres sont des valeurs par défaut.
    '''
    def __init__ (self):
        # Paramètres de discrétisation temporelle
        self.u0 = 1.
        self.a = 1.
        self.T = 1.
        self.cfl = 0.
        self.dt = 0.
        self.n = 0
        self.t = np.zeros(self.n + 1)
        
        # Tableaux des solutions exacte et approchée        
        self.solution_exacte = np.zeros(self.n + 1)
        self.solution_approchee = np.zeros(self.n + 1)
        
        # Fonctions python par défaut pour les simulations        
        self.function = linear
        self.exact = sol_ex_lin
        self.schema = EulerExplicite
```

```python
def linear(u, param) :
    return - param.a * u
```

```python
def quadra(u, param) :
    return - param.a * u**2
```

```python
def EulerExplicite(param) :
    u0 = param.u0
    dt = param.dt
    n = param.n
    
    u = np.zeros(n + 1)
    u[0] = u0
    for i in range(0, n) :
        u[i+1] = u[i] + dt * param.function(u[i], param) 
    param.solution_approchee = u
```

```python
def fonct_implicite(u_np1, u_n):
    g_u = param.function(u_np1, param) - (u_np1 - u_n) / param.dt
    return g_u
```

```python
def EulerImplicite(param) :
    u0 = param.u0
    dt = param.dt
    n = param.n
    u = np.zeros(n+1)
    u[0] = u0
    for i in range(0,n) :
        u[i+1] = scipy.optimize.fsolve(fonct_implicite, u[i], u[i])
    param.solution_approchee = u
```

```python
def tabtemps(param) :
    t=np.zeros(param.n+1)
    for i in range(0,param.n+1) :
        t[i]=i*param.dt
    param.t = t
```

```python
def calcul_solution_exacte(param) :
    param.solution_exacte = np.zeros(param.n + 1)
    param.solution_exacte = param.exact(param.t, param)
```

```python
def sol_ex_lin(t, param) :
    uex_t = np.exp(-param.a * t) * param.u0
    return uex_t
```

```python
def sol_ex_quad(t,param) :
    uex_t = param.u0 / (1 + param.u0 * param.a * t)
    return uex_t
```

```python
def graphique(param) :
    fig, ax = plt.subplots(figsize=(7,4))
    ax.set_xlim(-0.01, param.T + 0.1)

    ax.plot(param.t, param.solution_exacte, color='red', label='solution exacte ')
    ax.plot(param.t, param.solution_approchee, color='black', marker='*', lw=0, label='solution approchee ')
    plt.xlabel('Temps')
    plt.ylabel('Solution u(t)')
    ax.grid(True)
    ax.legend(loc='best')

    plt.draw()
```

## Déterminer numériquement l'ordre de convergence d'une méthode

### Euler explicite
On se propose d'étudier l'ordre de convergence numérique de la méthode d'**Euler explicite** implémentée la semaine dernière, à partir de 8 simulations successives pour lesquelles on multipliera à chaque fois le nombre d'itérations (= nombre de pas de temps) par 2 :   
* la simulation n°1 sera faite avec $n = 4$ itérations ($n=2^2$), jusqu'à un temps final $T=1$
* la simulation n°2 sera faite avec $n = 8$ itérations ($n=2^3$), jusqu'à un temps final $T=1$
* la simulation n°3 sera faite avec $n = 16$ itérations ($n=2^4$), jusqu'à un temps final $T=1$
* ...
* la simulation n°8 sera faite avec $n = 512$ itérations ($n=2^9$), jusqu'à un temps final $T=1$

**1)** Ecrire le code qui permet de créer un tableau `ite` qui contiendra les nombres d'itérations de chacune des 8 simulations (i.e : `ite = [4, 8, 16, 32, ..., 512]`) et un tableau `err` qui contiendra les erreurs de chacune des 8 simulations.

**Rappel** : On rappelle que pour une simulation, l'erreur est définie comme étant la valeur absolue de la différence entre la solution numérique au temps final obtenue avec Euler explicite et la solution exacte au temps final :    
Ex : pour la $i^{ème}$ simulation, on a $n=2^{i+1}$ et :
$$ err_i : \vert u_{num}(n \Delta t_i) - u_{ex}(n \Delta t_i) \vert, \quad \text{où} \quad \Delta t_i = \frac{T}{n} $$
**Indication** : 
* Les tableaux `ite` et `err` seront initialisés avec des 0 et seront mis à jour au sein d'une boucle.
* On ne jouera pas sur la CFL dans cette étude de convergence : le nombre de pas de temps $n$ sera mis à jour dans une boucle et le pas de temps $\Delta t$ sera mis à jour en fonction.

```python
# Création et remplissage des tableaux "ite" et "err" pour Euler explicite :
param = Param()

S = 8 # nb de simulations à faire dans l'étude convergence
tab_ite = np.zeros(S)
tab_err = np.zeros(S)
print(tab_ite, tab_err)

param. T = 1

for i in range (0,S):
    param.n = 2 ** (i+2)
    param.dt = param.T / param.n
    tab_ite[i] = param.n
    EulerExplicite(param)
    tabtemps(param)
    calcul_solution_exacte(param)
    tab_err[i] = np.abs(param.solution_exacte[-1] - \
                        param.solution_approchee[-1])
print(tab_ite, tab_err)
```

**2)** En vous servant de la question précédente et de la fonction Python `PlotConv` donnée ci-dessous, tracer la courbe de convergence de la méthode d'Euler explicite.

```python
def PlotConv(ite, err) :
    '''
    Fonction permettant de tracer le graphique de la convergence
    d'une methode numerique et donc de voir son ordre.
    @ ite : tableau numpy des nombres d'iterations
    @ err : tableau numpy des erreurs 
    '''
    
    # calcul des logarithmes base 2 des tableaux ite et err
    ite = np.log(ite) / np.log(2)
    err = np.log(err) / np.log(2)
    
    # tracé de la courbe de convergence
    fig, ax = plt.subplots(figsize=(7,7))
    ax.set_title( 'convergence' ) 
    ax.set_xlabel('logarithme base 2 du nombre d iterations')
    ax.set_ylabel('logarithme base 2 de l erreur')
    ax.plot (ite,err,'r*')   
    plt.grid(True)
    plt.draw()
    
    return
```

```python
# Appel de la fonction PlotConv à partir des tableaux "ite" et "err" :
PlotConv(tab_ite, tab_err)
```

**3)** Observer la fonction Python `OrdreConv` implémentée ci-dessous qui à partir d'un tableau `ite` et d'un tableau `err` (de même taille) renvoie l'ordre de convergence $p$ d'une méthode, calculé à partir de la formule vue dans la partie cours :
$$ \boxed{
p = \vert \alpha \vert = \big \vert \dfrac{{\rm{Cov}}(X,Y)}{{\rm{Var}}(X)} \big \vert}
$$

⚠ La difficulté ici est la bonne utilisation de la fonction `np.cov`.  

**Rappel** : Soient deux jeux de données $X$ et $Y$, la matrice de covariance associée est une matrice carrée, de taille $2 \times 2$ et symétrique. L'appel de la fonction `np.cov` renvoie donc toute la matrice de covariance, de taille $2 \times 2$. Ce que l'on souhaite ici, c'est le coefficient de corrélation entre $X$ et $Y$. Or, les coefficients sur la diagonale de la matrice de corrélation correspondent respectivement à Var($X$) et Var($Y$) et les deux coefficients non-diagonaux (qui sont égaux puisque la matrice est symétrique) correspondent au coefficient de corrélation entre $X$ et $Y$ $\longrightarrow$ C'est cette valeur qui nous intéresse ici.   
ATTENTION : ici l'option `bias=True` est à rajouter dans les arguments de la fonction `np.cov` (voir la documentation officielle).

```python
def OrdreConv(ite, err) :
    '''
    Fonction permettant de calculer l ordre d une methode
    par calcul du coefficient directeur de la droite de regression
    @ ite : tableau numpy des nombres d'iterations
    @ err : tableau numpy des erreurs
    '''
    
    # calcul des logarithmes base 2 des tableaux ite et err
    ite = np.log(ite) / np.log(2)
    err = np.log(err) / np.log(2)
    
    # calcul du coefficient directeur alpha
    x2m = np.var(ite)
    alpha = np.cov(ite, err, bias=True)[0,1] / x2m
    
    return np.abs(alpha)
```

```python
# Appel de la fonction OrdreConv à partir des tableaux "ite" et "err" :
OrdreConv(tab_ite, tab_err)
```
