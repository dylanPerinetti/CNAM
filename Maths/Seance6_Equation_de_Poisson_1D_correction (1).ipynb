## I - Equation de diffusion et équation de Poisson

### L'équation de diffusion : généralités et construction

Physiquement, la **diffusion** est un phénomène qui a la particularité d'être non réversible dans le temps.   
Ex : soient deux fluides miscibles, la diffusion d'un fluide dans l'autre tend vers un mélange homogène. 

Soit $\vec{u}$ la quantité physique ou la substance qui se diffuse. L'équation de diffusion consiste à décrire la concentration de $\vec{u}$ en chaque temps $t>0$ et en chaque point $\boldsymbol{x}$ du domaine spatial $\Omega \in \mathbb{R^n}$.

Pour établir cette équation de diffusion, on part tout d'abord de l'hypothèse que la quantité $u$ est conservée, c'est à dire qu'une fois générée par une source, elle n'est ni détruite ni dupliquée. Cela se traduit par une **équation de conservation** faisant intervenir l'opérateur différentiel de la **divergence** (voir cours du premier semestre sur les opérateurs différentiels) :

$$ \dfrac{\partial \vec{u}}{\partial t} + \text{div}(\vec{j}) = 0$$

où $\vec{j}$ est un courant généré naturellement par la différence de concentration spatiale de $\vec{u}$ (c'est le moteur du phénomène de diffusion).  
Ce courant $\vec{j}$ s'exprime comme suit :
$$\vec{j} = -\kappa \nabla \vec{u}$$

On voit qu'il est proportionnel au gradient de la concentration $\vec{u}$, avec le signe $-$ qui indique la direction du courant dans le domaine $\Omega$ orienté, et avec le coefficient de proportionnalité $\kappa$ qui correspond au coefficient de diffusion. Ce coefficient $\kappa$ dépend de la quantité diffusante $\vec{u}$, du milieu traversé et de leur interaction (viscosité, ...) 

Ainsi, en remplaçant l'expression de $\vec{j}$ dans la première équation on a :

$$ \dfrac{\partial \vec{u}}{\partial t} -\kappa \ \text{div}(\nabla \vec{u}) = 0$$
$$ \Longleftrightarrow \dfrac{\partial \vec{u}}{\partial t} -\kappa \Delta \vec{u} = 0 \quad {\text{(Rappel :}} \ \text{div}(\nabla \vec{u}) = \Delta \vec{u})$$

où "$\Delta$" est l'opérateur **Laplacien** (voir cours du premier semestre sur les opérateurs différentiels).

Le phénomène de diffusion peut être causé ou accentué par une "action" ou force extérieure. Celle-ci se matérialise mathématiquement par l'ajout d'un terme source $F$ dans le second membre de l'équation :

$$ \dfrac{\partial \vec{u}}{\partial t} -\kappa \Delta \vec{u} = F$$

Dans ce cours, comme pour celui sur l'équation d'advection, nous nous concentrerons sur le **cas 1D** où le domaine spatial $\Omega = [0,L]$ est un sous-ensemble de $\mathbb{R}$ (1 dimension). Dans ce cas, l'équation de diffusion s'écrit alors :

$$ (1) \quad \boxed{\dfrac{\partial u(x, t)}{\partial t} -\kappa \dfrac{\partial^2 u(x, t)}{\partial x^2} = F(x)} \qquad {\text{avec}}\quad t>0 \in \mathbb{R}, \ 0\leq x \leq L \in \mathbb{R}$$

### L'équation de Poisson : équation de diffusion en temps long

Dans le cadre de ce cours, on s’intéresse au comportement en temps long, c’est à dire pour $t \rightarrow +\infty$, de la solution $u(x, t)$. Nous admettons ici que l’évolution est de plus en plus lente, et donc que la dérivée
partielle $\dfrac{\partial u(x,t)}{\partial t}$ tend vers zéro et que la solution $u(x, t)$ converge vers une fonction limite $u_{\infty}(x)$ qui dépend seulement de l’espace. 

L’équation satisfaite par la fonction limite $u_{\infty}(x)$ consiste à éliminer la dérivée temporelle $\dfrac{\partial u(x,t)}{\partial t}$ de l’équation $(1)$. On trouve alors :

$$-\kappa \dfrac{\partial^2 u_{\infty}(x)}{\partial x^2} = F(x), \qquad {\text{avec}}\quad t>0 \in \mathbb{R}, \ 0\leq x \leq L $$

La solution limite au temps long satisfait à une équation aux dérivées partielles qui ne dépend que de l’espace et que l'on appelle **équation de Poisson** pour le Laplacien (du nom du mathématicien et physicien français Siméon Denis Poisson).   
On pose alors $F ≡ \kappa f$ et on remplace la dérivée partielle par rapport à $x$ par une dérivée ordinaire puisque maintenant le problème ne dépend que de l’espace. Il vient alors l'EDO suivante :
$$ (2) \qquad \boxed{-\dfrac{d^2u}{dx^2} = f(x)}, \qquad {\text{avec}}\quad t>0 \in \mathbb{R}, \ 0\leq x \leq L$$

#### Quelques exemples d'équations de Poisson (dans $\mathbb{R^d}$)
* Une des équations de Poisson les plus célèbres est l'**équation de la chaleur en régime stationnaire** vérifiée par la température $T \in \mathbb{R}^d$ :
$$-\Delta T = -\displaystyle \sum_{j=1}^d \dfrac{d^2T}{dx_j^2}= \dfrac{\sigma}{C}$$
où $\sigma \in \mathbb{R}^d$ est la source thermique et $C$ est la conductivité thermique du milieu.
* Un autre exemple d'équation de Poisson est celle de **Poisson-Boltzmann** vérifiée par le potentiel électrique $V$ :
$$-\Delta V = -\displaystyle \sum_{j=1}^d \dfrac{d^2V}{dx_j^2}= \dfrac{\rho}{\varepsilon}$$
où $\rho$ est la densité volumique de charge électrique associée à $V$ et $\varepsilon$ la permittivité électrique du milieu.

### Fermeture du problème par des conditions aux limites
Pour que l'équation de Poisson $(2)$ puisse être résolue (au même titre que toute EDO ou toute EDP impliquant des dérivées spatiales), il est nécessaire de lui associer des **conditions limites**, c'est à dire de prescrire des contraintes sur les valeurs que prend la solution $u(x)$ sur les bords/limites du domaine $\Omega=[0,L]$. 

Il existe plusieurs familles de conditions limites. Parmi les plus usuelles on citera les suivantes (exprimées dans le contexte 1D présent) :
* la condition de bord **périodique** (vue lors du TP sur l'équation d'advection) : 
$$u(L)=u(0)$$
* la condition de bord de type **Dirichlet** (nommée d’après Johann Dirichlet) : ce type de condition de bord consiste à directement spécifier les **valeurs** que la solution $u$ doit vérifier sur les frontières du domaine : 
$$u(0) = \alpha \quad \text{et} \quad u(L)= \beta$$
* la condition de bord de type **Neumann** (nommée d'après Carl Neumann) : ce type de condition de bord consiste à spécifier les **valeurs des dérivées** que la solution $u$ doit vérifier sur les frontières du domaine :
$$u'(0) = \alpha \quad \text{et} \quad u'(L)= \beta$$
* la condition de bord de type **Robin** (nommée d'après Victor Gustave Robin) : ce type de condition de bord est une **combinaison pondérée** d'une condition aux limites de **Dirichlet** et d'une condition aux limites de **Neumann** :
$$au(0)-bu'(0)=g(0)$$
$$au(L)+bu'(L)=g(L)$$
où $a, b$ et $g$ sont des fonctions définies sur le bord du domaine $\Omega=[0,L]$.

$\longrightarrow$ Dans ce cours, nous choisirons une condition limite de type **Dirichlet homogène**, c'est à dire $u(0) = 0$ et $u(L)= 0$, pour résoudre l'équation de Poisson $(2)$. On obtient alors le **problème de Poisson** suivant :

$$(3) \quad \left \{
\begin{array}{rcl}
&-\dfrac{d^2u}{dx^2} = f(x)   \\
&u(0) = 0 \quad \text{et} \quad u(L)= 0   \\
\end{array}
\right. 
$$ 

On peut montrer que ce problème possède au plus une solution.

## II - Discrétisation du problème de Poisson 1D

On cherche à résoudre le problème $(3)$ :
$$(3) \quad \left \{
\begin{array}{rcl}
&-\dfrac{d^2u}{dx^2} = f(x)   \\
&u(0) = 0 \quad \text{et} \quad u(L)= 0   \\
\end{array}
\right. 
$$ 
où $f$ est une fonction continue.

Pour la discrétisation du problème (qui n'est que spatiale ici), le maillage est construit en introduisant $N+1$ points de grilles (aussi appelés "noeuds") que l'on note $x_i$ avec $i= 0,1, \cdots, N$, régulièrement espacés avec un pas $\Delta x$. 
On note $u_i$ la valeur de la fonction solution $u(x)$ au noeud $x_i$.   
L'équation à résoudre s'écrit sous forme discrète en chaque noeud $x_i$ :

$$-\big(\dfrac{d^2u}{dx^2}\big)_i = f(x_i)=f_i$$

et les conditions aux limites de type Dirichlet homogène s’écrivent naturellement au niveau discret comme ceci :
$$u_0=0 \quad \text{et} \quad u_{N}=0.$$


Il nous reste donc $N-1$ inconnues scalaires $u_1,u_2, \cdots, u_{N-1}$ à déterminer.  
On approche la solution $u$ par la méthode des **différences finies**. Cette méthode va nous permettre ici d'approcher la dérivée seconde de $u$ à l'aide d'un schéma à plusieurs points, basé sur la formule de Taylor-Young. Ici on utilisera un schéma à 3 points, d'ordre 2 construit comme ceci :

En un point $x$ du domaine et pour une valeur $\Delta x$ du pas de discrétisation tels que $u$ soit trois fois dérivable sur l’intervalle $[x, x + 2\Delta x]$, la formule de Taylor-Young conduit aux 2 relations suivantes, obtenues respectivement avec un décentrage amont et aval :

$$\frac {u(x-\Delta x)-u(x)}{\Delta x}=-u'(x)+{\frac {\Delta x}{2}}u''(x)-\frac {\Delta x^{2}}{3!}u^{(3)}(x)-\Delta x^{2}\varepsilon_{1}(x,\Delta x)$$

$$\frac {u(x+\Delta x)-u(x)}{\Delta x}=u'(x)+{\frac {\Delta x}{2}}u''(x)+\frac {\Delta x^{2}}{3!}u^{(3)}(x)+\Delta x^{2}\varepsilon_{2}(x,\Delta x)$$

où les fonction $\varepsilon_{i}(x,\Delta x)$ convergent vers 0 avec $\Delta x$. Par conséquent, en sommant les deux relations ci-dessus on obtient :

$${\frac {u(x-\Delta x)-2u(x)+u(x+\Delta x)}{\Delta x^{2}}} \approx u''(x)$$

qui est une approximation de $u''(x)$ du 2e ordre en $\Delta x$ et qui s'écrit comme ceci aux noeuds du maillage :
$$\big(\dfrac{d^2u}{dx^2}\big)_i \approx {\frac {u_{i-1}-2u_i+u_{i+1}}{\Delta x^{2}}}, \qquad 1 \leq i \leq N-1$$


L'équation de Poisson discrétisée par la méthode des différences finies est alors donnée par :
$$\boxed{\frac {-u_{i-1}+2u_i-u_{i+1}}{\Delta x^{2}} = f_i} \ , \qquad 1 \leq i \leq N-1$$

En détaillant les $N+1$ équations qui constituent l'ensemble du problème, incluant les conditions de bord de type Dirichlet $u_0=\alpha$ et $u_N=\beta$, on obtient le système :
$$\left \{
\begin{array}{rcl}
u_0 &=& \alpha  \\
\frac{1}{\Delta_x^2}(-u_0 + 2u_1 -u_2) &=& f_1  \\
\frac{1}{\Delta_x^2}(-u_1 + 2u_2 -u_3) &=& f_2 \\
\cdots \\
\frac{1}{\Delta_x^2}(-u_{i-1} + 2u_i -u_{i+1}) &=& f_i \\
\cdots \\
\frac{1}{\Delta_x^2}(-u_{N-3} + 2u_{N-2} -u_{N-1}) &=& f_{N-2} \\
\frac{1}{\Delta_x^2}(-u_{N-2} + 2u_{N-1} -u_{N}) &=& f_{N-1} \\
u_N &=& \beta  \\
\end{array}
\right. \quad \Longleftrightarrow \quad \left \{
\begin{array}{rcl}
u_0 &=& \alpha  \\
\frac{1}{\Delta_x^2}(2u_1 -u_2) &=& f_1 + \alpha \ /\Delta x^2 \\
\frac{1}{\Delta_x^2}(-u_1 + 2u_2 -u_3) &=& f_2 \\
\cdots \\
\frac{1}{\Delta_x^2}(-u_{i-1} + 2u_i -u_{i+1}) &=& f_i \\
\cdots \\
\frac{1}{\Delta_x^2}(-u_{N-3} + 2u_{N-2} -u_{N}) &=& f_{N-2} \\
\frac{1}{\Delta_x^2}(-u_{N-2} + 2u_{N-1}) &=& f_{N-1} + \beta \ /\Delta x^2 \\
u_N &=& \beta  \\
\end{array}
\right.
$$

Ce système peut alors s'écrire sous la forme matricielle suivante $AU = F$ :

$$(4) \quad \underbrace{{\begin{pmatrix}1&0&0&0&0&\cdots&0&0\\0&2&-1&0&0&\cdots&0&0\\0&-1&2&-1&0&\cdots&0&0\\\vdots&0&\ddots&\ddots&\ddots&\ddots&\vdots&\vdots\\\vdots&\vdots&\ddots&\ddots&\ddots&\ddots&0&\vdots\\0&\cdots&0&0&-1&2&-1&0\\0&\cdots&0&0&0&-1&2&0\\0&\cdots&0&0&0&0&0&1\end{pmatrix}}}_{A \ \in \ \mathcal{M}_{N+1,N+1}} \quad \underbrace{{\begin{pmatrix}u_0\\u_1\\u_2\\\vdots\\\vdots\\u_{N-2}\\u_{N-1}\\u_N\end{pmatrix}}}_{U \ \in \ \mathbb{R}^{N+1}} = \underbrace{{\begin{pmatrix}\alpha\\\Delta x^2 f_1+\alpha \\\Delta x^2 f_2\\\vdots\\\vdots\\\Delta x^2 f_{N-2}\\\Delta x^2 f_{N-1}+\beta \\\beta\end{pmatrix}}}_{F \ \in \ \mathbb{R}^{N+1}}$$

qui, dans le cas de conditions de **Dirichlet homogènes** (i.e. $\alpha = \beta = 0$), s'écrit :

$$(4H) \quad{\begin{pmatrix}1&0&0&0&0&\cdots&0&0\\0&2&-1&0&0&\cdots&0&0\\0&-1&2&-1&0&\cdots&0&0\\\vdots&0&\ddots&\ddots&\ddots&\ddots&\vdots&\vdots\\\vdots&\vdots&\ddots&\ddots&\ddots&\ddots&0&\vdots\\0&\cdots&0&0&-1&2&-1&0\\0&\cdots&0&0&0&-1&2&0\\0&\cdots&0&0&0&0&0&1\end{pmatrix}} \quad {\begin{pmatrix}u_0\\u_1\\u_2\\\vdots\\\vdots\\u_{N-2}\\u_{N-1}\\u_N\end{pmatrix}} = \Delta x^2 \ {\begin{pmatrix}0\\f_1\\f_2\\\vdots\\\vdots\\f_{N-2}\\f_{N-1}\\0\end{pmatrix}}$$
 
La matrice $A$ est une matrice à coefficients réels  
* symétrique (i.e. $A = A^T$ )
* tridiagonale (i.e. tous les coefficients qui ne sont ni sur la diagonale principale, ni sur la sous-diagonale et sur-diagonale sont nuls)  

$\longrightarrow$ C’est une matrice inversible.

**Conclusion** : Pour résoudre le problème de Poisson par la méthode des différences finies, on doit résoudre un système linéaire, c-a-d inverser la matrice $A$.

* **cas 2** : $f(x)=\dfrac{k^2 \pi^2}{L_x^2} \sin\big(\frac{k\pi x}{L_x} \big)$. On peut alors démontrer que la solution exacte est égale à $u(x) =\sin\big(\frac{k\pi x}{L_x} \big).$ On prendra $k=1$ ici.

**1)** La classe `Param` et la fonction habituelle `tabx` sont données ci-dessous.

```python
# Import des packages python
%matplotlib notebook
%matplotlib inline
import time
import numpy as np
import scipy.linalg as sp_linalg
import scipy.optimize as sp_optimize
import matplotlib.pyplot as plt
import matplotlib.cm as cm
```

```python
class Param (object) : 
    def __init__ (self):
        self.type_rhs = 1 # numéro du cas (1 ou 2) pour le choix du second membre f
        self.Lx = 1 # longueur du domaine
        self.nx = 16 # nombre de cellules dans le maillage (nombre entier)
        self.dx = 0. # pas d'espace
        self.x = np.zeros(self.nx + 1) # tableau des coordonées x dans le maillage (x_0 = 0, x_1 = dx, ..., x_N = Lx)
```

```python
def tabx(param) :
    param.x = np.linspace(0, param.Lx, param.nx + 1)
```

On donne aussi ci-dessous la fonction python `sol_exacte` qui renvoie en sortie le vecteur $U_{ex} \in \mathbb{R}^{N+1}$ contenant la solution exacte du problème de Poisson $(3)$ aux points $i$ du maillage, selon que l'on se trouve dans le cas 1 ou 2.

Dans cette fonction on se sert de l'attribut `type_rhs` de `param` : 
* si `type_rhs` est égal à 1 la solution exacte sera celle du cas 1, 
* sinon si `type_rhs` est égal à 2 la solution exacte sera celle du cas 2, 
* sinon le programme renvoie un message d'erreur (on pourra pour cela utiliser l'instruction : `raise ValueError("Message")`)

```python
def sol_exacte(param) :
    Lx = param.Lx
    x = param.x
    u_ex = np.zeros(param.nx + 1)
    if param.type_rhs == 1 :
        u_ex = 2. * x * (Lx - x)
    elif param.type_rhs == 2 :
        u_ex = np.sin(np.pi * x / Lx)
    else :
        raise ValueError("Attention, valeur pour type_rhs inexistante ! (choisir 1 ou 2)")
    return u_ex
```

**2)** Compléter la fonction python `assembleRHS` ci-dessous qui renvoie en sortie le vecteur $F \in \mathbb{R}^{N+1}$ second membre de l'équation matricielle $(4)$, selon que l'on se trouve dans le cas 1 ou 2 (on pourra de même se servir de l'attribut `type_rhs` de `param`).

**Note** : *rhs* vient de "right hand side"= second membre en anglais

```python
def assembleRHS(param) :
    Lx = param.Lx
    x = param.x
    rhs = np.zeros(param.nx + 1)
    if param.type_rhs == 1 :
        rhs[1:-1] = 4
        #rhs[1:-1] = rhs[1:-1] + 4
        #rhs[1:-1] = np.ones(param.nx - 1) * 4
    elif param.type_rhs == 2 :
        rhs[1:-1] = (np.pi / Lx) ** 2 * np.sin((np.pi * x[1:-1])/ Lx)
        # ou
        #rhs = (np.pi / Lx) ** 2 * np.sin((np.pi * x)/ Lx)
        #rhs[0], rhs[-1] = 0, 0
    else :
        raise ValueError("Attention, valeur pour type_rhs inexistante ! (choisir 1 ou 2)")
    return rhs * (param.dx ** 2)
```

**3)** Compléter la fonction python `assembleMat` ci-dessous qui renvoie en sortie la matrice $A \in \mathcal{M}_{N+1,N+1}$ de l'équation matricielle $(4)$ (i.e matrice du problème de Poisson discrétisé avec conditions de Dirichlet homogènes).

```python
def assembleMat(param) :
    A = np.zeros((param.nx + 1, param.nx + 1))
    for i in range(1, param.nx - 1):
        A[i, i] = 2 # diagonale
        A[i + 1, i] = -1 # sous-diagonale
        A[i, i + 1] = -1 # sur-diagonale
    A[0, 0] = 1
    A[-1, -1] = 1
    A[-2, -2] = 2
    return A
```

```python
param = Param()
param.nx = 10
assembleMat(param)
```

Soit la fonction `isoplot1d` donnée ci-dessous (complète) permettant de tracer sur un même graphique la solution exacte et la solution approchée du problème de Poisson :

```python
def isoplot1d(u, u_ex, param) :
    ''' Fonction pour l'affichage graphique des solutions 1D '''
    # Setup figures
    fig, ax = plt.subplots(figsize=(6,6))
    ax.set_xlim(-.01, param.Lx + 0.01)
    #if (param.type_rhs == 1) :
    #    ax.set_ylim(-.01, .51  )
    if (param.type_rhs == 2) :
        ax.set_ylim(-1.15, 1.15 )
    ax.set_title('probleme de Poisson Dirichlet en 1D pour cas '+str(param.type_rhs)) 
    ax.set_xlabel('x')
    absx = np.linspace(0, param.Lx, param.nx+1)

    # Tracé des courbes des solutions exacte et approchee :
    couleur = 'blue'
    if (param.type_rhs <= 2):
        # Stockage de la premiere ligne de la solution exacte dans un vecteur 1D 
        uuex = np.zeros(param.nx+1)
        uuex[0:param.nx+1] = u_ex[0:param.nx+1]    
        ax.plot (absx, uuex, couleur, label='solution exacte')

    j = 0 
    uu = np.zeros(param.nx + 1)
    uu[0:param.nx+1]= u[j*(param.nx+1):(j+1)*(param.nx+1)]    
    ax.plot (absx, uu, color='red', marker='*' , lw=0, label='solution approchee  ')
    ax.legend(loc='lower center')
    ax.grid(True)

    return
```

**4)** Compléter la fonction python ci-dessous `main_Poisson` : il s'agit du programme principal ("main program" en anglais) dont le rôle est d'appeler toutes les étapes nécessaires à la résolution numérique du problème de Poisson $(4)$.

*Indication* : Pour la résolution du système linéaire $AU = F$ on pourra utiliser la fonction `np.linalg.solve` de numpy (voir la documentation officielle pour son utilisation).

```python
def main_Poisson(param):
    """
    Programme principal pour la résolution numérique du problème de Poisson :
    -d2u / dx2 = f(x) 
    avec u(0)=0 et u(Lx)=0
    """
    # Définition du pas d'espace et les coordonnées x_i des points du maillage
    param.dx = param.Lx / param.nx
    tabx(param)

    # Calcul de la solution exacte (si disponible)
    sol_ex = sol_exacte(param)

    # Assemblage plein de la matrice 
    A = assembleMat(param)
    
    # Assemblage du vecteur second membre (RHS)
    RHS = assembleRHS(param)

    # Resolution du systeme lineaire AU = F, matrice pleine (voir fonction python np.linalg.solve)
    sol_approchee = np.linalg.solve(A, RHS) # sol_approchee = U
        
    # Représentation 1D de la solution
    isoplot1d(sol_approchee, sol_ex, param)
    
    return sol_approchee
```

```python
def main_Poisson(param):
    """
    Programme principal pour la résolution numérique du problème de Poisson :
    -d2u / dx2 = f(x) 
    avec u(0)=0 et u(Lx)=0
    Calcul d'erreur inclu à la fin.
    """
    # Définition du pas d'espace et les coordonnées x_i des points du maillage
    param.dx = param.Lx / param.nx
    tabx(param)

    # Calcul de la solution exacte (si disponible)
    sol_ex = sol_exacte(param)

    # Assemblage plein de la matrice 
    A = assembleMat(param)
    
    # Assemblage du vecteur second membre (RHS)
    RHS = assembleRHS(param)

    # Resolution du systeme lineaire AU = F, matrice pleine (voir fonction python np.linalg.solve)
    sol_approchee = np.linalg.solve(A, RHS) # sol_approchee = U
        
    # Représentation 1D de la solution
    isoplot1d(sol_approchee, sol_ex, param)

    # Calcul de l'erreur L_infini si solution exacte disponible
    pas = np.log(param.nx) / np.log(2)
    err = -200
    if (param.type_rhs <= 2) : 
        errsolu = np.max(np.abs(sol_approchee - sol_ex))
        print ('erreur si solution exacte ' , errsolu ) 
        err = np.log (errsolu) / np.log(2) 
    
    return pas, err
```

```python
def OrdreConv(pas, err) :
    '''
    Entrées : 
    @pas : tableau qui contient les log_2(nx)
    @err : tableau qui contient les log_2(err)
    
    Sortie : 
    @alpha : coefficient directeur de la droite 
    de convergence = ordre de convergence numérique
    '''
    xm = np.mean(pas)        
    ym = np.mean(err)        
      
    x2m = np.var(pas)
    y2m = np.var(err)
    alpha = np.cov(pas,err,bias=True)[0,1] / x2m
    return np.abs(alpha)
```

```python
def PlotConv(pas,err) :
    # Graphique de la convergence 
    '''
    Entrées : 
    @pas : tableau qui contient les log_2(nx)
    @err : tableau qui contient les log_2(err)
    '''
    fig, ax = plt.subplots(figsize=(6,6))
    ax.set_title('Convergence' ) 
    ax.set_xlabel('logarithme base 2 du nb de points du maillage')
    ax.set_ylabel('logarithme base 2 de l erreur')
    ax.plot (pas, err,'r*')   
    plt.draw()
    return
```

```python
param = Param()

# Initialisation des paramètres
S = 5
tab_err = np.zeros(S) 
tab_nx = np.zeros(S) 

param.type_rhs = 2
param.nx = 4

# Boucle sur les maillages dans laquelle on appelle le programme principal "main_Poisson" 
for i in range(0, S):
    #param.nx = 2 ** (i + 2)
    tab_nx[i], tab_err[i] = main_Poisson(param) # simu n°i
    param.nx = param.nx * 2 # ou param.nx = 2 ** (i + 3) 

# Calcul de l'ordre de convergence et tracé de la courbe de convergence
ordre = OrdreConv(tab_nx, tab_err)
print('Ordre de convergence :', ordre)
PlotConv(tab_nx, tab_err)
```

## Annexe : détail du calcul des solutions exactes
**Résolution analytique du problème de Poisson** :
$$  -\dfrac{d^2u}{dx^2} = f(x) \qquad \qquad \qquad (3)$$
$$ u(0) = 0 \quad \text{et} \quad u(L)= 0 \qquad $$

#### Cas 1 : $f(x)=4$ 
L'équation à résoudre est une EDO d'ordre 2, linéaire et à coefficients constants :

$$(E_1) \qquad -\dfrac{d^2u}{dx^2} = 4$$

* Equation homogène associée : $-\dfrac{d^2u}{dx^2} = 0$.  
L'équation caratéristique de l'équation homogène associée est $r^2=0$, donc la solution est la racine double $r=0$.   
Donc la solution de l'équation homogène est de la forme $u_h(x) = (\lambda_1 + \lambda_2 x)e^{rt}$. Ainsi :
$$u_h(x) = (\lambda_1 + \lambda_2 x), \quad \lambda_1, \lambda_2 \in \mathbb{R}$$


* Solution particulière (méthode de similarité) :  
Ici le second membre est de la forme $f(x)=P(x)$ où $P$ est un polynôme de degré $0$.    
On cherche une solution particulière $u_{p}(x)$ et un polynôme $Q$ de degré $0$, c'est à dire une constante $C$, avec :

    * $u_{p}(x)=C$ si 0 n'est pas racine de l'équation caractéristique
    * $u_{p}(x)=Cx$ si 0 est racine simple de l'équation caractéristique
    * $u_{p}(x)=Cx^{2}$ si 0 est racine double de l'équation caractéristique  
On se trouve ici dans le 3ème cas. Donc la solution particulière est de la forme $u_{p}(x)=Cx^{2}$ et par identification avec l'équation d'origine on trouve :
$$u_{p}(x)=-2x^{2} \qquad (C=2)$$
* Par superposition on a donc que la solution générale est égale à :
$$u(x) = (\lambda_1 + \lambda_2 x) - 2x^2$$

* En utilisant les conditions de type Dirichlet homogènes sur les bords du domaine, $u(0)=0$ et $u(L_x)=0$, on trouve :
$$\lambda_1=0, \quad \lambda_2=2L_x$$
La solution exacte et unique de l'équation $(E_1)$ est :
$$\boxed{u(x) = 2xL_x  - 2x^2 = 2x(L_x  - x)}$$

#### Cas 2 : $f(x)=\dfrac{k^2 \pi^2}{L_x^2} \ \sin\big(\frac{k\pi x}{L_x} \big)$ 
L'équation à résoudre est l'EDO d'ordre 2 suivante :

$$(E_2) \qquad -\dfrac{d^2u}{dx^2} = \dfrac{k^2 \pi^2}{L_x^2} \ \sin\big(\frac{k\pi x}{L_x} \big)$$

Avec des conditions de bords de type Dirichlet homogène, on peut utiliser la décomposition en série de Fourier en sinus pour résoudre l'équation.

**Rappel** : La décomposition d'une fonction $u(x)$ en série de Fourier en sinus est définie par :
$$u(x) : \displaystyle \sum_{k=1}^{\infty} \widehat{u}_k \sin\big(\frac{k\pi x}{L_x} \big)$$
$$ \text{avec} \ \ \widehat{u}_k = \dfrac{2}{L_x} \displaystyle \int_0^{L_x} u(x) \sin\big(\frac{k\pi x}{L_x} \big) \ dx$$

Donc $$-\dfrac{d^2u}{dx^2}= f(x) = \displaystyle \sum_{k=1}^{\infty} \dfrac{k^2 \pi^2}{L_x^2} \widehat{u}_k \sin\big(\frac{k\pi x}{L_x} \big)$$
$$ \text{avec} \ \ \widehat{u}_k = \dfrac{\dfrac{2}{L_x}}{\dfrac{k^2 \pi^2}{L_x^2}} \displaystyle \int_0^{L_x} f(x) \sin\big(\frac{k\pi x}{L_x} \big) \ dx$$

or $f(x)=\dfrac{k^2 \pi^2}{L_x^2} \sin\big(\frac{k\pi x}{L_x} \big)$ donc :
$$\widehat{u}_k = \dfrac{\dfrac{2}{L_x}}{\dfrac{k^2 \pi^2}{L_x^2}} \displaystyle \int_0^{L_x} \dfrac{k^2 \pi^2}{L_x^2} \ \sin\big(\frac{k\pi x}{L_x} \big)\sin\big(\frac{k\pi x}{L_x} \big) \ dx$$
$$ \Longleftrightarrow \dfrac{2}{L_x} \displaystyle \int_0^{L_x} \sin\big(\frac{k\pi x}{L_x} \big)\sin\big(\frac{k\pi x}{L_x} \big) \ dx$$
$$ \Longleftrightarrow \boxed{u(x) = \sin\big(\frac{k\pi x}{L_x} \big)}$$
