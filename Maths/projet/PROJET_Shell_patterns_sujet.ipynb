Ce projet repose sur les travaux de recherche de **Hans Meinhardt** qui s'est intéressé à "la beauté algorithmique des coquillages". Ses travaux ont fait l'objet d'un ouvrage : _Meinhardt, H. (2009). The algorithmic beauty of sea shells. Springer Science & Business Media_.

Comme expliqué par H. Meinhardt dans son livre, les motifs pigmentaires des coquillages tropicaux présentent une grande beauté et une grande diversité, avec un fort mélange de régularité et d’irrégularité. Un motif particulier semble obéir à des règles spécifiques, mais ces règles autorisent des variations. Aucun coquillage n’est identique à un autre. Les motifs immobiles paraissent statiques et, en effet, ils sont constitués de matière calcifiée. Cependant, le mécanisme sous-jacent qui engendre ces motifs est entièrement **dynamique**.

Un mollusque ne peut agrandir sa coquille qu’au niveau du bord de croissance (cf lignes bleues sur les Figure 2(a) et 2(b) ci-dessous). Le plus souvent, c'est à ce bord que de nouveaux éléments du motif pigmentaire sont ajoutés.  

Comment fonctionne le dépôt de pigment sur le bord de croissance d'un coquillage ? Comme le schématise la Figure 1(a), le dépôt de pigment est sous le contrôle d’une substance chimique appelée **activateur**, qui stimule sa propre production par un mécanisme de rétroaction positive appelé  **autocatalyse** (boucle de rétroaction verte sur le schéma). Pour qu’un motif puisse se former sur la coquille, un autre mécanisme est également nécessaire afin de supprimer la production de l’activateur dans le voisinage des centres autocatalytiques : c'est le mécanisme d'**inhibition**. Celui-ci empêche l’activateur de se propager sur l’ensemble du bord de croissance. Le motif résulte ainsi de l’interaction antagoniste entre une activation à courte portée (on dit que l'activateur présente une faible diffusion) et une inhibition à longue portée (on dit que l'inhibiteur présente une forte diffusion).

> **Figure 1 - Formation de motifs par autocatalyse et inhibition à longue portée** _(source : Meinhardt, H. (2009))_     
(a) Schéma réactionnel : un activateur catalyse sa propre production ainsi que celle de son antagoniste à forte diffusion, l’inhibiteur.  
(b–e) 4 étapes de la formation du motif au cours du temps après une perturbation locale. Simulation numérique sur un réseau linéaire de cellules (toutes les cellules, en abcisses, constituent les cellules du bord de croissance). On voit que la distribution homogène des deux substances est instable. En effet, une très faible augmentation locale de l’activateur (en vert) croît progressivement jusqu’à atteindre un état stationnaire (e) dans lequel l’auto-activation et le nuage environnant d’inhibiteur (en rouge) s’équilibrent.

## Modèle 1 : Le modèle activateur-inhibiteur
L'interaction entre l'activateur autocatalytique $a(x,t)$ (noté $a$ ci-dessous) et son antagoniste, l'inhibiteur $b(x,t)$ (noté $b$ ci-dessous), est donnée par le système d'EDPs suivant :

$$
\begin{array}{rclr} \dfrac{\partial a}{\partial t} &=& s \left ( \dfrac{a^2}{b} + b_a\right) - r_a a + D_a \dfrac{\partial ^2 a}{\partial x^2} & \qquad (1.a) \\
\dfrac{\partial b}{\partial t} &=& s a^2 + b_b - r_b b + D_b \dfrac{\partial ^2 b}{\partial x^2} & \qquad (1.b)
\end{array}
$$

**Variables** (= inconnues) :  
* $ a= a(x,t)$ : **activateur autocatalytique** (= substance chimique déclenchant le dépôt de pigment)
* $ b= b(x,t)$ : **inhibiteur** (= substance chimique freinant localement la production de l'activateur afin d’empêcher son extension incontrôlée et de structurer ainsi le motif spatial).

Avec les conditions initiales (au temps $t=0$) :$
\begin{array}{rcl}a(x,0) &=& a_0(x) > 0 \\
b(x,0) &=& b_0(x) > 0\\
\end{array}
$

**Paramètres** (= données du problème) :
* Constantes en temps et en espace :
    * $b_a, b_b$ : petite **source externe de production** de l'activateur et de l'inhibiteur, respectivement
    * $r_a, r_b$ : **taux de décroissance** de l'activateur et de l'inhibiteur, respectivement
    * $D_a, D_b$ : **coefficient de diffusion** de l'activateur et de l'inhibiteur, respectivement.   
* Constantes en temps :
    * $s = s(x)$ : **densité de source** : décrit la capacité des cellules à réaliser l’autocatalyse.  
    
$\rightarrow$ _Exemple_ : La Figure 2 ci-dessous montre deux exemples de mollusques (i.e. _Cepea nemoralis_ et _Lyria planicostata taiwanica_) dont la formation du motif en lignes droites est pilotée par ce modèle activateur-inhibiteur.

> **Figure 2 - Formation de motifs périodiques stables dans l’espace.** _(source : Meinhardt, H. (2009))_     
(a) Coquille de _Cepea nemoralis_ et (b) coquille de _Lyria planicostata taiwanica_ (le bord de croissance est marqué par la ligne bleue).  
(c) Simulation numérique sur un réseau linéaire de cellules (réparties en absisses (x) - le bord de croissance est marqué par la ligne bleue). Les concentrations de l’activateur (en haut) et de l’inhibiteur (en bas) sont représentées en fonction du temps. Comme la taille du domaine (c.a.d du bord de croissance) est supposée beaucoup plus grande que la portée de diffusion de l’inhibiteur, plusieurs maxima se forment. En raison de fluctuations aléatoires, l’espacement est un peu irrégulier. Des maxima de l'activateur peuvent apparaître trop proches les uns des autres et ne pas survivre à l’inhibition mutuelle. Toutefois, une distance minimale et une distance maximale sont maintenues.  
(d) Simulation similaire avec une représentation espace-temps analogue à celle observée sur les coquillages.

## Modèle 2 : Le modèle activateur-substrat
Le mécanisme d'inhibition n’est pas nécessairement dû à la présence d'une molécule inhibitrice distincte (comme dans le modèle 1). Il peut aussi émerger simplement du fait que la ressource nécessaire à la production de pigment est localement consommée plus vite qu’elle n’est régénérée. Cette ressource est appelée le "**substrat**" et peut-être vue comme une molécule nutritive ou métabolique.

Dans ce contexte, la formation de motifs à la surface des coquillages peut-être modélisée par le modèle "activateur-substrat" dans lequel l’activateur stimule sa propre production, mais cette production consomme le substrat dont l’épuisement local crée alors une inhibition indirecte.

L'interaction entre l'activateur autocatalytique $a(x,t)$ (noté $a$ ci-dessous) et le substrat $b(x,t)$ (noté $b$ ci-dessous), est donnée par le système d'EDPs suivant :

$$
\begin{array}{rclr} \dfrac{\partial a}{\partial t} &=& s \cdot b \cdot \tilde{a}^2 - r_a a + D_a \dfrac{\partial ^2 a}{\partial x^2} & \qquad (2.a) \\
\dfrac{\partial b}{\partial t} &=& \sigma - s \cdot b \cdot \tilde{a}^2 - r_b b + D_b \dfrac{\partial ^2 b}{\partial x^2} & \qquad (2.b)
\end{array}
$$

où cette interaction est employée usuellement en imposant une condition de saturation sur l'activateur via le coefficient de saturation $s_a$ dans la définition de $\tilde{a}^2$ :

$$ \tilde{a}^2 = \dfrac{a^2}{1 + s_a a^2} + b_a$$ 

**Variables** (= inconnues) :  
* $ a= a(x,t)$ : **activateur autocatalytique** (= substance chimique déclenchant le dépôt de pigment)
* $ b= b(x,t)$ : **substrat** (= ressource biochimique nécessaire à la production de pigment).

Avec les conditions initiales (au temps $t=0$) :$
\begin{array}{rcl}a(x,0) &=& a_0(x) > 0 \\
b(x,0) &=& b_0(x) > 0\\
\end{array}
$

**Paramètres** (= données du problème) :
* Constantes en temps et en espace :
    * $b_a$ : petite **source externe de production** de l'activateur
    * $r_a, r_b$ : **taux de décroissance** de l'activateur et du substrat, respectivement
    * $D_a, D_b$ : **coefficient de diffusion** de l'activateur et du substrat, respectivement
    * $s_a$ : coefficient de saturation de l'activateur
* Constantes en temps :
    * $s = s(x)$ : **densité de source** : décrit la capacité des cellules à réaliser l’autocatalyse.
    * $\sigma = \sigma(x)$ : **taux de production du substrat**.
    
    
$\rightarrow$ _Exemple_ : La Figure 3(a) ci-dessous montre un exemple de mollusque (i.e. _Amoria undulata_) dont la formation du motif en lignes ondulantes est pilotée par ce modèle activateur-substrat, avec une source de production de substrat $\sigma(x)$ _périodique_ le long du bord de croissance.

## Modèle 3 : Le modèle activateur-inhibiteur-hormone

Pour les motifs présents sur certains coquillages, une observation empirique suggère l’existence d’un mécanisme de contrôle global qui surveille la quantité totale d’activateur dans le système et déclenche de nouvelles ondes lorsque sa concentration devient trop faible. Ce processus est régulé globalement par une **hormone** $c$ qui contrôle la quantité totale d’activateur le long du bord en croissance.

L'interaction entre l'activateur autocatalytique $a(x,t)$ (noté $a$ ci-dessous), l'inhibiteur $b(x,t)$ (noté $b$ ci-dessous) et l'hormone $c(t)$ (notée $c$ ci-dessous) est donnée par le système d'EDPs suivant :

$$
\begin{array}{rclr} \dfrac{\partial a}{\partial t} &=& \dfrac{s \cdot a^2}{1+ s_a a^2}\cdot \dfrac{1}{b_b + b} - r_a a + D_a \dfrac{\partial ^2 a}{\partial x^2} & \qquad (3.a) \\
\dfrac{\partial b}{\partial t} &=& \dfrac{s \cdot a^2}{1+ s_a a^2} - \dfrac{r_b b}{c} + D_b \dfrac{\partial ^2 b}{\partial x^2} & \qquad (3.b) \\
\dfrac{\partial c}{\partial t} &=& -r_c c + \dfrac{1}{L_x} \displaystyle \int_0^{L_x} (r_c a) \ dx & \qquad (3.c)
\end{array}
$$

Les deux premières équations $(3.a)$ et $(3.b)$ représentent un système activateur-inhibiteur (où on retrouve, en plus du modèle 1, la condition de saturation sur l'activateur via le coefficient de saturation $s_a$). En revanche ici, contrairement au modèle 1, celui-ci nous montre qu'une faible concentration d'activateur (i.e. faible nombre d'ondes en lignes obliques) entraîne une faible concentration de l’hormone, ce qui accélère la décroissance de l’inhibiteur. La concentration de l’activateur augmente alors et atteint, en certains points, le seuil à partir duquel de nouvelles ondes se forment. Il s’agit d’un processus autorégulé, dans lequel l’hormone assure une rétroaction négative, maintenant ainsi le nombre d’ondes obliques à un niveau approximativement constant.

**Variables** (= inconnues) :  
* $ a= a(x,t)$ : **activateur autocatalytique** 
* $ b= b(x,t)$ : **inhibiteur** 
* $ c= c(t)$ : **hormone** (la concentration de l'hormone est supposée constante en espace le long du bord de croissance.)

Avec les conditions initiales (au temps $t=0$) :$
\begin{array}{lcl}a(x,0) &=& a_0(x) \geq 0 \\
b(x,0) &=& b_0(x) > 0\\
c(0) &=& c_0 > 0\\
\end{array}
$


**Paramètres** (= données du problème) :
* Constantes en temps et en espace :
    * $b_b$ : petite **source externe de production** de l'inhibiteur
    * $r_a, r_b, r_c$ : **taux de décroissance** de l'activateur, de l'inhibiteur et de l'hormone, respectivement
    * $D_a, D_b$ : **coefficient de diffusion** de l'activateur et du substrat, respectivement
    * $s_a$ : coefficient de saturation de l'activateur
* Constantes en temps :
    * $s = s(x)$ : **densité de source** : décrit la capacité des cellules à réaliser l’autocatalyse.
    
$\rightarrow$ _Exemple_ : La Figure 3(b) ci-dessus montre un exemple de mollusque (i.e. _Oliva Porphyria_) dont la formation d'ondes en collision à l'origine du motif en "$>$" est pilotée par ce modèle activateur-inhibiteur-hormone.

(a(x,t), b(x,t)) & \mapsto& (f_1(x,t),f_2(x,t)) = (\underbrace{s \left ( \dfrac{a^2}{b} + b_a\right) - r_a a + D_a \dfrac{\partial ^2 a}{\partial x^2}}_{f_1}, \ \underbrace{s a^2 + b_b - r_b b + D_b \dfrac{\partial ^2 b}{\partial x^2}}_{f_2})
\end{array}
