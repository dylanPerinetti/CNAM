# Partie A : Compréhension du problème

### Modèle 1 : activateur-inhibiteur

**1)** Expliquez avec vos mots la signification des deux équations $(1.a)-(1.b)$ du modèle "activateur-inhibiteur"...

**Réponse** :

**Équation (1.a) — activateur $a$ :**
$$\frac{\partial a}{\partial t} = s\left(\frac{a^2}{b} + b_a\right) - r_a a + D_a \frac{\partial^2 a}{\partial x^2}$$

- **Autocatalyse** : le terme $s \cdot a^2/b$ représente l'auto-catalyse de l'activateur. L'activateur stimule sa propre production (proportionnellement à $a^2$), mais cette production est **réduite par l'inhibiteur** $b$ (division par $b$ : quand $b$ augmente, la production de $a$ diminue).
- **Source externe** : $s \cdot b_a$ est une petite source externe constante de production de l'activateur.
- **Décroissance** : $-r_a a$ représente la dégradation naturelle de l'activateur (proportionnelle à sa concentration).
- **Diffusion spatiale** : $D_a \partial^2 a/\partial x^2$ modélise la diffusion de l'activateur dans l'espace.

**Équation (1.b) — inhibiteur $b$ :**
$$\frac{\partial b}{\partial t} = s a^2 + b_b - r_b b + D_b \frac{\partial^2 b}{\partial x^2}$$

- **Catalyse de $b$ par $a$** : le terme $s \cdot a^2$ représente la production de l'inhibiteur stimulée par l'activateur (l'activateur « fabrique » son propre antagoniste).
- **Source externe** : $b_b$ est une petite source externe de production de l'inhibiteur.
- **Décroissance** : $-r_b b$ représente la dégradation naturelle de l'inhibiteur.
- **Diffusion spatiale** : $D_b \partial^2 b/\partial x^2$ modélise la diffusion (plus forte que celle de $a$) de l'inhibiteur.

**Identification des phénomènes fondamentaux de la Figure 1 :**
| Phénomène | Terme responsable |
|-----------|------------------|
| Auto-catalyse de l'activateur | $s \cdot a^2/b$ dans (1.a) |
| Catalyse de l'inhibiteur par l'activateur | $s \cdot a^2$ dans (1.b) |
| Réduction de la production de $a$ par $b$ | Division par $b$ dans $s \cdot a^2/b$ de (1.a) |
| Diffusion spatiale de $a$ | $D_a \partial^2 a/\partial x^2$ dans (1.a) |
| Diffusion spatiale de $b$ | $D_b \partial^2 b/\partial x^2$ dans (1.b) |

**2)** Identifiez les deux coefficients du modèle impliqués dans l'activation à courte portée / inhibition à longue portée.

**Réponse** :

Les deux coefficients impliqués sont les **coefficients de diffusion** $D_a$ (diffusion de l'activateur) et $D_b$ (diffusion de l'inhibiteur).

- **Activation à courte portée** : $D_a$ petit → l'activateur diffuse peu, son influence reste locale.
- **Inhibition à longue portée** : $D_b$ grand → l'inhibiteur diffuse loin, son action s'étend sur de grandes distances.

**Relation nécessaire :** $\boxed{D_a \ll D_b}$

Cette inégalité est la condition fondamentale de l'instabilité de Turing : un état homogène stable sans diffusion peut devenir instable et générer des motifs spatiaux dès lors que les deux espèces diffusent avec des vitesses très différentes.

### Modèle 2 : activateur-substrat

**3)** Identifiez dans les équations $(2.a)-(2.b)$ les termes modélisant les phénomènes en gras...

**Réponse** :

Termes du modèle activateur-substrat :

- **"l'activateur stimule sa propre production"** : terme $s \cdot b \cdot \tilde{a}^2$ dans (2.a), avec $\tilde{a}^2 = \dfrac{a^2}{1+s_a a^2} + b_a$. Ce terme est positif et croissant avec $a$ : l'activateur stimule sa propre production (autocatalyse), avec saturation via $s_a$.

- **"cette production consomme le substrat"** : terme $-s \cdot b \cdot \tilde{a}^2$ dans (2.b). C'est exactement le même terme qu'en (2.a), mais avec un signe **négatif** : la même réaction qui produit de l'activateur consomme du substrat $b$.

- **"l'épuisement local crée une inhibition indirecte"** : quand le substrat $b$ diminue localement (consommation > régénération $\sigma$), le terme $s \cdot b \cdot \tilde{a}^2$ dans (2.a) diminue (car il est proportionnel à $b$), ce qui **freine la production de $a$**. L'inhibition est donc indirecte, via l'épuisement de la ressource.

**Différences avec le modèle 1 :**
| Aspect | Modèle 1 | Modèle 2 |
|--------|----------|----------|
| Mécanisme d'inhibition | Inhibiteur distinct $b$ actif | Épuisement indirect du substrat $b$ |
| Saturation | Aucune | $\tilde{a}^2 = a^2/(1+s_a a^2)+b_a$ |
| Régénération | $b_b$ constant | $\sigma(x)$ variable en espace |
| Rôle de $b$ | Antagoniste direct | Ressource consommée |

### Modèle 3 : activateur-inhibiteur-hormone

**4)** Identifiez dans les équations $(3.a)-(3.b)-(3.c)$ les termes modélisant les phénomènes en gras...

**Réponse** :

- **"faible concentration d'activateur → faible concentration de l'hormone"** : dans (3.c), $\dfrac{\partial c}{\partial t} = -r_c c + \dfrac{1}{L_x}\int_0^{L_x} r_c a\,dx$. Si $a$ est faible partout, l'intégrale $\int r_c a\,dx$ est petite, donc $\partial c/\partial t \approx -r_c c < 0$ → $c$ **diminue**.

- **"ce qui accélère la décroissance de l'inhibiteur"** : dans (3.b), le terme de décroissance de $b$ est $-\dfrac{r_b b}{c}$. Quand $c$ est petit, le rapport $r_b/c$ est **grand** → la décroissance de $b$ est **accélérée**.

- **"la concentration de l'activateur augmente alors"** : dans (3.a), le terme autocatalytique vaut $\dfrac{s a^2}{1+s_a a^2} \cdot \dfrac{1}{b_b + b}$. Quand $b$ diminue (suite à sa décroissance accélérée), le dénominateur $(b_b + b)$ diminue → ce terme **augmente** → $a$ croît jusqu'au seuil de formation d'une nouvelle onde.

# Partie B : Discrétisation du problème

**4)** Les fonctions `tabtemps` et `tabx` ci-dessous sont déjà implémentées. Quel est leur rôle respectif ?

**Réponse** :
- `tabtemps(param)` : construit le tableau `param.t` des $n_t+1$ instants discrets régulièrement espacés de $\Delta t=1$, soit $t[i] = i \cdot \Delta t$ pour $i \in [0, n_t]$. Ce tableau représente la discrétisation temporelle du problème.
- `tabx(param)` : construit le tableau `param.x` des $n_x+1$ positions discrètes régulièrement espacées de $\Delta x=1$, soit $x[i] = i \cdot \Delta x$ pour $i \in [0, n_x]$. Ce tableau représente la discrétisation spatiale du bord de croissance.

### Import des packages et définitions

```python
%matplotlib inline
import time
import numpy as np
import scipy.optimize
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import warnings
from IPython import display
warnings.filterwarnings('ignore')
```

La fonction `diffusion_periodique` implémente la discrétisation de l'opérateur de diffusion 1D $\partial^2 q/\partial x^2$ avec conditions aux limites **périodiques** (via `np.roll`). Avec $\Delta x = 1$, elle calcule simplement $q[i-1] - 2q[i] + q[i+1]$ (différences finies centrées d'ordre 2).

```python
def diffusion_periodique(q, param):
    """Approximation de d²q/dx² par différences finies centrées, CL périodiques, dx=1."""
    return (np.roll(q, -1) - 2*q + np.roll(q, 1)) * param.dx**2
```

Fonctions utilitaires pour construire les tableaux d'espace et de temps :

```python
def tabtemps(param):
    t = np.zeros(param.nt + 1)
    for i in range(0, param.nt + 1):
        t[i] = i * param.dt
    param.t = t

def tabx(param):
    x = np.zeros(param.nx + 1)
    for i in range(0, param.nx + 1):
        x[i] = i * param.dx
    param.x = x
```

### Question B.2 — Implémentation des 3 modèles (second membres $f$)

```python
def modele_activateur_inhibiteur(param):
    """
    Modèle 1 — second membre f = (f1, f2).
    Lit la solution courante dans param.solution_courante (shape: nx+1 x 2).
    Retourne f de même shape.
    """
    a = param.solution_courante[:, 0]   # activateur
    b = param.solution_courante[:, 1]   # inhibiteur

    # Protection contre la division par zéro
    b_safe = np.maximum(b, 1e-10)

    f = np.zeros((param.nx + 1, 2))
    # f1 : équation de l'activateur
    f[:, 0] = (param.s * (a**2 / b_safe + param.ba)
               - param.ra * a
               + param.Da * diffusion_periodique(a, param))
    # f2 : équation de l'inhibiteur
    f[:, 1] = (param.s * a**2 + param.bb
               - param.rb * b
               + param.Db * diffusion_periodique(b, param))
    return f
```

```python
def modele_activateur_substrat(param):
    """
    Modèle 2 — second membre f = (f1, f2).
    Utilise le coefficient de saturation sa et le taux de substrat sigma(x).
    """
    a = param.solution_courante[:, 0]   # activateur
    b = param.solution_courante[:, 1]   # substrat

    # ã² avec condition de saturation
    a_tilde_sq = a**2 / (1 + param.sa * a**2) + param.ba

    f = np.zeros((param.nx + 1, 2))
    # f1 : équation de l'activateur
    f[:, 0] = (param.s * b * a_tilde_sq
               - param.ra * a
               + param.Da * diffusion_periodique(a, param))
    # f2 : équation du substrat
    f[:, 1] = (param.sigma
               - param.s * b * a_tilde_sq
               - param.rb * b
               + param.Db * diffusion_periodique(b, param))
    return f
```

```python
def modele_activateur_inhibiteur_hormone(param):
    """
    Modèle 3 — second membre f = (f1, f2, f3).
    La hormone c est constante en espace → stockée dans param.solution_courante[:, 2]
    (toutes les lignes ont la même valeur).
    """
    a = param.solution_courante[:, 0]          # activateur
    b = param.solution_courante[:, 1]          # inhibiteur
    c = max(param.solution_courante[0, 2], 1e-10)  # hormone (scalaire, constante en x)

    # Terme autocatalytique commun
    autocatal = param.s * a**2 / (1 + param.sa * a**2)

    f = np.zeros((param.nx + 1, 3))
    # f1 : équation de l'activateur
    f[:, 0] = (autocatal / (param.bb + b)
               - param.ra * a
               + param.Da * diffusion_periodique(a, param))
    # f2 : équation de l'inhibiteur
    f[:, 1] = (autocatal
               - param.rb * b / c
               + param.Db * diffusion_periodique(b, param))
    # f3 : équation de l'hormone (scalaire → même valeur sur tous les points)
    # ∫(rc*a)dx / Lx ≈ (1/nx) * sum(rc*a)  [car dx=1, Lx=nx]
    f3_val = -param.rc * c + (1.0 / param.nx) * np.sum(param.rc * a)
    f[:, 2] = f3_val
    return f
```

### Question B.3 — Schémas en temps : Euler explicite et Heun

Ces fonctions effectuent **un seul pas de temps** : elles lisent `solution_courante` (= $U^n$) et écrivent $U^{n+1}$ dans `solution_courante`.

$$\text{Euler explicite :}\quad U^{n+1} = U^n + \Delta t\, f(U^n)$$

$$\text{Heun :}\quad \widetilde{U}_1 = U^n + \Delta t\,f(U^n),\quad \widetilde{U}_2 = U^n + \Delta t\,f(\widetilde{U}_1),\quad U^{n+1} = \tfrac{1}{2}(\widetilde{U}_1 + \widetilde{U}_2)$$

```python
def euler_explicite(param):
    """Un pas de temps Euler explicite (ordre 1). Très court par construction."""
    f = param.modele(param)
    param.solution_courante[:, :] = param.solution_courante[:, :] + param.dt * f
```

```python
def heun(param):
    """Un pas de temps Heun (ordre 2, predictor-corrector explicite)."""
    # Sauvegarde de U^n
    U_n = param.solution_courante.copy()

    # Prédicteur : Ũ₁ = U^n + dt*f(U^n)
    f_Un = param.modele(param)
    U_tilde_1 = U_n + param.dt * f_Un

    # Correcteur : Ũ₂ = U^n + dt*f(Ũ₁)
    param.solution_courante[:, :] = U_tilde_1
    f_Ut1 = param.modele(param)
    U_tilde_2 = U_n + param.dt * f_Ut1

    # Mise à jour : U^{n+1} = (Ũ₁ + Ũ₂)/2
    param.solution_courante[:, :] = 0.5 * (U_tilde_1 + U_tilde_2)
```

### Question B.1 — Classe `Param` complétée

La classe `Param` regroupe **tous** les attributs nécessaires aux simulations :
- Paramètres de discrétisation (déjà donnés)
- Paramètres physiques des modèles ($r_a, r_b, D_a, D_b$, etc.)
- Valeurs initiales ($a_0, b_0, c_0$)
- Tableaux de source $s(x)$ et $\sigma(x)$ (remplis par `initialisation`)
- Pointeurs de fonctions (`modele`, `schema`)

`solution_courante` et `solution_prec` sont des **matrices** de taille $(n_x+1) \times d_d$ où $d_d=2$ pour les modèles 1 et 2, $d_d=3$ pour le modèle 3. Chaque ligne correspond à un point spatial, chaque colonne à une variable ($a$, $b$, et éventuellement $c$).

```python
class Param(object):
    """
    Classe regroupant tous les paramètres et tableaux de la simulation.
    Les valeurs par défaut correspondent au modèle 1, cas 1.
    """
    def __init__(self):
        # --- Discrétisation spatiale et temporelle ---
        self.dd  = 2          # dimension du vecteur solution (2 ou 3)
        self.T   = 0.
        self.dt  = 1.         # dt = 1 (cf. énoncé : nt = T, dt = 1)
        self.nt  = 0
        self.t   = np.zeros(self.nt + 1)
        self.L   = 0
        self.dx  = 1.         # dx = 1 (cf. énoncé : nx = Lx, dx = 1)
        self.nx  = 0
        self.x   = np.zeros(self.nx + 1)

        # Tableaux solutions : (nx+1) lignes × dd colonnes
        self.solution_courante = np.zeros((self.nx + 1, self.dd))
        self.solution_prec     = np.zeros((self.nx + 1, self.dd))

        # Stockage temporel de l'activateur a(x,t) : (nt+1) × (nx+1)
        self.stocke_a = np.zeros((self.nt + 1, self.nx + 1))

        # --- Paramètres physiques des modèles ---
        self.ra        = 0.05   # taux de décroissance de l'activateur
        self.rb        = 0.08   # taux de décroissance de l'inhibiteur / substrat
        self.rc        = 0.     # taux de décroissance de l'hormone (modèle 3)
        self.ba        = 0.     # source externe de l'activateur
        self.bb        = 0.     # source externe de l'inhibiteur
        self.Da        = 0.001  # coefficient de diffusion de l'activateur
        self.Db        = 0.40   # coefficient de diffusion de l'inhibiteur
        self.sa        = 0.     # coefficient de saturation de l'activateur (mod. 2 et 3)
        self.sigma_max = 0.     # taux max de production du substrat (mod. 2)

        # --- Valeurs initiales ---
        self.a0 = 1.50   # concentration initiale de l'activateur
        self.b0 = 1.50   # concentration initiale de l'inhibiteur / substrat
        self.c0 = 0.     # concentration initiale de l'hormone (modèle 3)

        # --- Tableaux de source (remplis par initialisation()) ---
        self.s     = np.zeros(self.nx + 1)   # densité de source s(x)
        self.sigma = np.zeros(self.nx + 1)   # taux de substrat sigma(x) (mod. 2)

        # --- Fonctions modèle et schéma numérique ---
        self.modele = modele_activateur_inhibiteur
        self.schema = euler_explicite
```

### Question B.5 — Fonction `initialisation`

```python
def initialisation(param):
    ## Appel de tabx et tabtemps pour construire les grilles
    tabx(param)
    tabtemps(param)

    ## === 1) s(x) : fluctuations aléatoires de l'autocatalyse le long du bord
    # s(x) = 0.08 * ra * rand(x) + 0.96 * ra
    param.s = 0.08 * param.ra * np.random.rand(param.nx + 1) + 0.96 * param.ra

    ## === 2) sigma(x) : fluctuation périodique de la source de substrat (modèle 2 seulement)
    if param.modele == modele_activateur_substrat:
        x_arr = np.arange(0, param.nx + 1)
        param.sigma = param.sigma_max * np.abs(np.sin(4 * np.pi * x_arr / param.nx))

    ## === Initialisation des tableaux solutions
    param.solution_prec     = np.zeros((param.nx + 1, param.dd))
    param.solution_courante = np.zeros((param.nx + 1, param.dd))

    # Conditions initiales homogènes
    param.solution_prec[:, 0] = param.a0
    param.solution_prec[:, 1] = param.b0
    if param.dd == 3:
        param.solution_prec[:, 2] = param.c0

    ## === Activation initiale de "a" de façon aléatoire sur quelques cellules (30 maximum)
    i = 1
    for j in range(30):
        param.solution_prec[i, 0] = 1
        i = i + np.random.randint(0, 101)
        if i > param.nx:
            break

    ## === A l'initialisation (t = 0) on impose : solution_courante = solution_prec
    param.solution_courante[:, :] = param.solution_prec[:, :]
```

### Question B.6 — Programme principal `main_run_exec`

```python
def main_run_exec(param):
    # --- Initialisation du problème ---
    initialisation(param)

    # Création du tableau de stockage de l'activateur a(x,t)
    param.stocke_a = np.zeros((param.nt + 1, param.nx + 1))
    param.stocke_a[0] = param.solution_courante[:, 0]

    # --- Boucle en temps ---
    for n in range(1, param.nt + 1):
        # Étape 1 : avancer d'un pas de temps (lit solution_courante = U^n, écrit U^{n+1})
        param.schema(param)

        # Étape 2 : mettre à jour solution_prec ← solution_courante (pour le pas suivant)
        param.solution_prec[:, :] = param.solution_courante[:, :]

        # Étape 3 : stocker l'activateur au temps courant
        param.stocke_a[n] = param.solution_courante[:, 0]

    # --- Tracé des profils spatiaux au temps final ---
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.set_xlim(0., param.L + 0.1)
    plt.plot(param.x, param.solution_courante[:, 0], color='green', lw=2., label='activateur')
    plt.plot(param.x, param.solution_courante[:, 1], color='red',   lw=2., label='inhibiteur')
    plt.xlabel('position (x)')
    plt.title(f'Profils spatiaux au temps final T = {param.T}')
    plt.legend(loc='best')
    plt.grid(True)
    plt.show()
```

```python
np.random.seed(0)  # reproductibilité

param1_euler = Param()
param1_euler.nx  = 100
param1_euler.nt  = 1000
param1_euler.dx  = 1.
param1_euler.dt  = 1.
param1_euler.L   = param1_euler.nx
param1_euler.T   = param1_euler.nt
# Paramètres physiques — cas 1
param1_euler.ra  = 0.05
param1_euler.rb  = 0.08
param1_euler.ba  = 0.00
param1_euler.bb  = 0.00
param1_euler.Da  = 0.001
param1_euler.Db  = 0.40
param1_euler.a0  = 1.50
param1_euler.b0  = 1.50
param1_euler.dd  = 2
param1_euler.modele = modele_activateur_inhibiteur
param1_euler.schema = euler_explicite

main_run_exec(param1_euler)

# --- Affichage du motif espace-temps ---
fig = plt.figure(figsize=(16, 5))
ax = fig.add_subplot(121, projection='3d')
T_mesh, X_mesh = np.meshgrid(param1_euler.t, param1_euler.x, indexing='ij')
ax.plot_surface(T_mesh, X_mesh, param1_euler.stocke_a, cmap=cm.coolwarm, antialiased=True)
ax.set_xlabel('temps (t)')
ax.set_ylabel('position (x)')
ax.set_title('Motif 3D — Modèle 1, Cas 1 (Euler)')

ax2 = fig.add_subplot(122)
ax2.contourf(T_mesh, X_mesh, param1_euler.stocke_a, levels=30, cmap=cm.binary)
ax2.set_xlabel('temps (t)')
ax2.set_ylabel('position (x)')
ax2.set_title('Contour — Modèle 1, Cas 1 (Euler)')
plt.tight_layout()
plt.show()
```

### Simulation modèle 1 — Cas 1 : avec Heun

```python
np.random.seed(0)  # même graine pour comparer

param1_heun = Param()
param1_heun.nx  = 100
param1_heun.nt  = 1000
param1_heun.dx  = 1.
param1_heun.dt  = 1.
param1_heun.L   = param1_heun.nx
param1_heun.T   = param1_heun.nt
param1_heun.ra  = 0.05
param1_heun.rb  = 0.08
param1_heun.ba  = 0.00
param1_heun.bb  = 0.00
param1_heun.Da  = 0.001
param1_heun.Db  = 0.40
param1_heun.a0  = 1.50
param1_heun.b0  = 1.50
param1_heun.dd  = 2
param1_heun.modele = modele_activateur_inhibiteur
param1_heun.schema = heun

main_run_exec(param1_heun)

fig = plt.figure(figsize=(16, 5))
ax = fig.add_subplot(121, projection='3d')
T_mesh, X_mesh = np.meshgrid(param1_heun.t, param1_heun.x, indexing='ij')
ax.plot_surface(T_mesh, X_mesh, param1_heun.stocke_a, cmap=cm.coolwarm, antialiased=True)
ax.set_xlabel('temps (t)')
ax.set_ylabel('position (x)')
ax.set_title('Motif 3D — Modèle 1, Cas 1 (Heun)')

ax2 = fig.add_subplot(122)
ax2.contourf(T_mesh, X_mesh, param1_heun.stocke_a, levels=30, cmap=cm.binary)
ax2.set_xlabel('temps (t)')
ax2.set_ylabel('position (x)')
ax2.set_title('Contour — Modèle 1, Cas 1 (Heun)')
plt.tight_layout()
plt.show()
```

**Comparaison Heun vs Euler explicite :**

Les deux schémas produisent **qualitativement le même motif** (mêmes positions des rayures, même dynamique). Les différences quantitatives sont minimes car :
- Les termes de réaction sont non-linéaires mais les paramètres $(r_a, r_b, D_a, D_b)$ sont choisis de telle sorte que le système est bien résolu avec $\Delta t = 1$.
- Heun est d'ordre 2 en $\Delta t$ ($\text{erreur} \sim \Delta t^2$) contre l'ordre 1 d'Euler explicite ($\text{erreur} \sim \Delta t$). À $\Delta t=1$, cet avantage est visible mais ne change pas la nature du motif.

Pour **réduire les différences**, on pourrait prendre un $\Delta t$ plus petit (ex. $\Delta t = 0.5$), ce qui améliorerait la précision des deux schémas, avec un gain relatif plus grand pour Heun.

### Bonus : Étude de convergence en temps (Euler explicite vs Heun)

On suit le même protocole que les séances : on fixe $T$ et $n_x$, on fait varier $n_t$ par doublement successif, et on utilise la résolution la plus fine comme solution de référence (à défaut de solution exacte).

On reporte les erreurs avec les **mêmes fonctions** `PlotConv` / `OrdreConv` / `PlotConv2schemas` que les séances.

```python
# ======== Fonctions de convergence (style séances) ========

def PlotConv(ite, err):
    """Tracé de la courbe de convergence en log base 2 (style séances)."""
    ite_log = np.log(ite) / np.log(2)
    err_log = np.log(err) / np.log(2)
    fig, ax = plt.subplots(figsize=(7, 7))
    ax.set_title('Courbe de convergence')
    ax.set_xlabel('logarithme base 2 du nombre de pas de temps')
    ax.set_ylabel('logarithme base 2 de l erreur')
    ax.plot(ite_log, err_log, 'r*')
    plt.grid(True)
    plt.draw()

def OrdreConv(ite, err):
    """Calcul de l'ordre de convergence par régression linéaire (style séances)."""
    ite_log = np.log(ite) / np.log(2)
    err_log = np.log(err) / np.log(2)
    x2m = np.var(ite_log)
    alpha = np.cov(ite_log, err_log, bias=True)[0, 1] / x2m
    return np.abs(alpha)

def PlotConv2schemas(ite, err_EE, err_Heun):
    """Comparaison des courbes de convergence pour 2 schémas (style PlotConv4c des séances)."""
    ite_log    = np.log(ite)      / np.log(2)
    err_EE_log = np.log(err_EE)   / np.log(2)
    err_H_log  = np.log(err_Heun) / np.log(2)

    fig, ax = plt.subplots(figsize=(7, 7))
    ax.set_title('Convergence en temps — Euler explicite vs Heun')
    ax.set_xlabel('logarithme base 2 du nombre de pas de temps')
    ax.set_ylabel('logarithme base 2 de l erreur')
    ax.plot(ite_log, err_EE_log, 'b*', label='Euler explicite')
    ax.plot(ite_log, err_H_log,  'r*', label='Heun')
    ax.legend(loc='best')
    plt.grid(True)
    plt.draw()

# ======== Simulation de convergence ========
# On fixe T et nx, on double nt à chaque fois (même principe que les séances).
# dt = T / nt varie → on sort du cadre dt=1 pour étudier la convergence.

def run_conv(schema_func, nt_val, T_sim=50., nx_sim=100, seed=42):
    """Simule le modèle 1 avec les paramètres du cas 1 et renvoie a au temps final."""
    np.random.seed(seed)
    pc = Param()
    pc.nx = nx_sim;  pc.nt = nt_val
    pc.dx = 1.;      pc.dt = T_sim / nt_val   # dt variable
    pc.L  = nx_sim;  pc.T  = T_sim
    pc.ra = 0.05;    pc.rb = 0.08
    pc.ba = 0.;      pc.bb = 0.
    pc.Da = 0.001;   pc.Db = 0.40
    pc.a0 = 1.50;    pc.b0 = 1.50
    pc.dd = 2
    pc.modele = modele_activateur_inhibiteur
    pc.schema = schema_func
    main_run_exec(pc)
    return pc.solution_courante[:, 0].copy()

# Nombre de simulations (même style que les séances : n0 * 2^k)
nb_sim = 5
T_sim  = 50.
nx_sim = 100
n0     = 50   # nombre de pas de temps initial

ite    = np.array([n0 * 2**k for k in range(nb_sim)], dtype=float)

# Solution de référence = résolution la plus fine (dernier ite)
a_ref_EE   = run_conv(euler_explicite, int(ite[-1]), T_sim, nx_sim)
a_ref_Heun = run_conv(heun,            int(ite[-1]), T_sim, nx_sim)

# Calcul des erreurs (norme euclidienne, comme en séance)
err_EE   = np.zeros(nb_sim - 1)
err_Heun = np.zeros(nb_sim - 1)
for k in range(nb_sim - 1):
    a_EE   = run_conv(euler_explicite, int(ite[k]), T_sim, nx_sim)
    a_H    = run_conv(heun,            int(ite[k]), T_sim, nx_sim)
    err_EE[k]   = np.linalg.norm(a_EE   - a_ref_EE)
    err_Heun[k] = np.linalg.norm(a_H    - a_ref_Heun)

# Tracé des courbes de convergence séparées (style séances)
PlotConv(ite[:-1], err_EE)
plt.gca().set_title('Convergence — Euler explicite')
print(f'Ordre Euler explicite : {OrdreConv(ite[:-1], err_EE):.2f}  (théorique : 1)')

PlotConv(ite[:-1], err_Heun)
plt.gca().set_title('Convergence — Heun')
print(f'Ordre Heun            : {OrdreConv(ite[:-1], err_Heun):.2f}  (théorique : 2)')

# Tracé comparatif (style PlotConv4c des séances)
PlotConv2schemas(ite[:-1], err_EE, err_Heun)
```

# Partie E : Simulations avec le modèle 3
## Motifs d'ondes en collision ($>$) — *Oliva porphyria*

**Paramètres :** $r_a=0.10$, $r_b=0.014$, $r_c=0.10$, $b_b=0.10$, $D_a=0.015$, $D_b=0.00$, $s_a=0.25$.

**Initialisation :** $a_0=0.00$, $b_0=0.10$, $c_0=0.10$, $n_x \sim \mathcal{O}(1000)$, $n_t \approx 5\,n_x$.

```python
np.random.seed(2)

param3 = Param()
param3.nx  = 1000
param3.nt  = 5000
param3.dx  = 1.; param3.dt  = 1.
param3.L   = param3.nx; param3.T = param3.nt
param3.ra  = 0.10
param3.rb  = 0.014
param3.rc  = 0.10
param3.ba  = 0.00
param3.bb  = 0.10
param3.Da  = 0.015
param3.Db  = 0.00
param3.sa  = 0.25
param3.a0  = 0.00; param3.b0 = 0.10; param3.c0 = 0.10
param3.dd  = 3
param3.modele = modele_activateur_inhibiteur_hormone
param3.schema = euler_explicite

main_run_exec(param3)

# Motif espace-temps
fig, axes = plt.subplots(1, 2, figsize=(18, 6))
T_m, X_m = np.meshgrid(param3.t, param3.x, indexing='ij')
axes[0].contourf(T_m, X_m, param3.stocke_a, levels=40, cmap=cm.binary)
axes[0].set_xlabel('temps (t)'); axes[0].set_ylabel('position (x)')
axes[0].set_title('Motif Oliva porphyria — Modèle 3 (activateur-inhibiteur-hormone)')
# Zoom pour voir les ">"
axes[1].contourf(T_m[2000:4000, :], X_m[2000:4000, :], param3.stocke_a[2000:4000, :],
                 levels=40, cmap=cm.binary)
axes[1].set_xlabel('temps (t)'); axes[1].set_ylabel('position (x)')
axes[1].set_title('Zoom — ondes en collision en ">"')
plt.tight_layout(); plt.show()

# Evolution de l'hormone c(t)
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(param3.t, param3.stocke_a.mean(axis=1), 'g-', lw=1.5, label='moyenne spatiale a(t)')
ax.set_xlabel('temps (t)'); ax.set_ylabel('concentration')
ax.set_title('Concentration moyenne de l\'activateur (proxy pour c(t))')
ax.legend(); ax.grid(True); plt.show()
```

**Analyse de la solution — Modèle 3 :**

Le modèle activateur-inhibiteur-hormone produit les **ondes en collision** caractéristiques de l'*Oliva porphyria* (motif en « $>$ »).

**Mécanisme dynamique :**
1. Des ondes d'activateur se forment à partir des perturbations initiales aléatoires et se propagent dans les deux directions (gauche et droite).
2. Quand deux ondes se rencontrent, elles **s'annihilent** (inhibition mutuelle), créant le motif en « $>$ ».
3. L'hormone $c$ joue un rôle de **régulateur global** : quand toutes les ondes s'éteignent simultanément (trop peu d'activateur), $c$ baisse, réduisant l'inhibition de $b$ ($r_b b / c$ augmente → $b$ chute rapidement), ce qui libère l'activateur et permet l'émergence de **nouvelles ondes**.
4. Ce processus est **auto-régulé** : l'hormone maintient un nombre approximativement constant d'ondes actives à tout instant.

**Paramètres clés :**
- $D_a = 0.015 \gg D_b = 0$ : l'activateur diffuse, pas l'inhibiteur (inhibition très locale).
- $s_a = 0.25$ : saturation modérée → les ondes ont une amplitude contrôlée.
- $b_b = 0.10$ : source externe d'inhibiteur qui empêche $b$ de tomber à zéro et stabilise le front des ondes.
